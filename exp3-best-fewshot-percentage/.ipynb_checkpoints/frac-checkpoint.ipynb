# Πείραμα 2 - Εύρεση καλύτερου ποσοστού εκπαίδευσης
#### Διπλωματική εργασία - Αναγνωστόπουλος Σπύρος

### 1) Πειγραφή πειράματος
Ξεκινώντας την εξειδικευμένη εκπαίδευση στα δεδομένα μετοχών, σε αυτό το πείραμα διερευνάμε πως αντιδρά το μοντέλο όσο μεγαλώνει το ποσοστό του συνόλου δεδομένων που δίνεται για εκπαίδευση. Για κάθε ποσοστό (6%, 12%, 18%, 24%, 30%, 36%, 42%, 48%, 54%), παρακολουθείται η πορεία των σφαλμάτων εκπαίδευσης (δεδομένα που έχει δει) και αξιολόγησης (δεδομένα που δεν έχει δει) σε πέντε εποχές. Επιπλέον, υπολογίζονται οι νόρμες του συνόλου των κλίσεων για κάθε εποχή και τέλος υπολογίζουμε το χάσμα γενίκευσης. Σκοπός μας είναι να χρησιμοποιήσουμε όσο το δυνατό λιγότερα δεδομένα.

#### Ερευνητικά Ερωτήματα Πειράματος 2 – Εύρεση καλύτερου ποσοστού εκπαίδευσης:

- Υπάρχουν ποσοστά που καταλήγουν σε μοντέλο με υψηλή ικανότητα γενίκευσης;

- Υπάρχουν περιπτώσεις ποσοστών που οδηγούν σε υπερπροσαρμογή;

- Πότε και πού το μοντέλο μαθαίνει δυναμικά, και πότε σταθεροποιείται;

- Ποιό ποσοστό παρουσιάζει την καλύτερη επίδοση με όσο το δυνατό μικρότερο αριθμό δεδομένων;

Στα διαγράμματα φαίνεται μια ξεκάθαρη εξέλιξη στη μαθησιακή συμπεριφορά του μοντέλου και στην ικανότητά του να γενικεύει σε μετοχές διαφορετικής μεταβλητότητας.

Στα χαμηλότερα ποσοστά (6%–18%), το μοντέλο αντιμετωπίζει δυσκολίες, παρουσιάζοντας υψηλό σφάλμα αξιολόγησης για πιο πολύπλοκες μετοχές (π.χ. TSLA, AMD), παρά το γεγονός ότι το σφάλμα εκπαίδευσης μειώνεται σταθερά και φτάνει σε πολύ χαμηλές τιμές. Αυτό είναι ισχυρή ένδειξη ότι το μοντέλο κάνει υπερπροσαρμογή (overfitting) στο σύνολο εκπαίδευσης. Παρόλα αυτά, ακόμα και με τόσο λίγα παραδείγματα, μετοχές χαμηλής μεταβλητότητας όπως οι PG, JNJ και XOM επιτυγχάνουν ήδη χαμηλό και σταθερό σφάλμα αξιολόγησης, υποδεικνύοντας ότι απλώς χρειαζόμαστε μικρή προσαρμογή του προεκπαιδευμένου μοντέλου για να εντοπιστούν τα μοτίβα τους.

Ένα σημείο καμπής παρατηρείται μετά το ποσοστό του 18%. Η TSLA, που αρχικά παρουσίαζε ασταθή απόδοση και υψηλό σφάλμα αξιολόγησης, παρουσιάζει σημαντική βελτίωση. Το σφάλμα αξιολόγησης μειώνεται απότομα και σταθεροποιείται. Αυτή η αλλαγή δείχνει ότι το μοντέλο αρχίζει να συλλαμβάνει τη δυναμική δομή της TSLA μόλις εκπαιδευτεί σε επαρκή αριθμό παραδειγμάτων.

Στα μεσαία ποσοστά (24%,30%,36%), οι περισσότερες μετοχές παρουσιάζουν ισορροπία μεταξύ μείωσης του σφάλματος εκπαίδευσης και βελτίωσης του σφάλματος αξιολόγησης. Το σφάλμα αξιολόγησης δείχνει πως η υπερπροσαρμογή μειώνεται. Οι TSLA, PG και SPY αποδίδουν σταθερά καλά σε αυτό το εύρος. H AMD φαίνεται ότι σταθεροποιείται στο 6 σε αυτό το στάδιο. Η AAPL σταθεροποιείτε λίγο πάνω από το 1.

Στα υψηλότερα ποσοστά (42%,48%,54%), το μοντέλο επιτυγχάνει τη σταθερότερη και πιο γενικεύσιμη απόδοσή του. Το σφάλμα αξιολόγησης παραμένει σταθερά χαμηλό τόσο για σταθερά όσο και για ευμετάβλητες μετοχές, ενώ το σφάλμα εκπαίδευσης συγκλίνει ομαλά με ελάχιστη διακύμανση. Όμως στα ποσοστά 48% και 54% το σφάλμα αξιολόγησης της AMD αποκτά ανοδική πορεία σε σχέση με το σφάλμα εκπαίδευσης αυτό σημαίνει πως ξαναεμφανίζεται υπερπροσαρμογή.

### 3) Νόρμα του συνόλου των κλίσεων

Όταν εκπαιδεύουμε ένα νευρωνικό δίκτυο, το μοντέλο μαθαίνει προσαρμόζοντας τις παραμέτρους του ώστε να μειώσει το σφάλμα. Αυτή η προσαρμογή γίνεται με βάση τις παράγωγους της εκάστοτε συνάρτησης κόστους ως προς κάθε μία από αυτές τις παραμέτρους. Η νόρμα του συνολικού μεγέθους των κλίσεων (gradient norm) είναι ένα συνολικό μέγεθος που δείχνει την αλλαγή που θα εφαρμοστεί ως προς όλες τις παραμέτρους.

Η νόρμα αυτή δείχνει το πόσο έντονα προσπαθεί να μάθει το μοντέλο από τα δεδομένα σε κάθε βήμα. Υψηλές τιμές νόρμας υποδηλώνουν ενεργή μαθησιακή διαδικασία, δηλαδή ότι το μοντέλο κάνει σημαντικές προσαρμογές. Αντίθετα, χαμηλές τιμές δείχνουν σύγκλιση. Στο πλαίσιο του πειράματος, η παρακολούθηση της νόρμας βοηθά στην κατανόηση της δυναμικής της μάθησης όσο το διαθέσιμο σύνολο δεδομένων αυξάνεται.

Αντί να κάνω γράφημα για κάθε μετοχή του συνόλου δεδομένων ξεχωριστά, παίρνω τον μέσο όρο τους για κάθε εποχή και κάνω ένα 3D γράφημα που δείχνει το κατά μέσο όρο ποσό της νόρμας για κάθε ποσοστό, σε κάθε εποχή.

Το τρισδιάστατο γράφημα επιφάνειας δείχνει ότι τα χαμηλά ποσοστά εκπαίδευσης συσχετίζονται με υψηλότερους και ασταθείς νόρμες,δηλαδή ασταθείς ενημερώσεις παραμέτρων και θορυβώδη βελτιστοποίηση.

Καθώς το ποσοστό δεδομένων εκπαίδευσης αυξάνεται, οι νόρμες μειώνονται και η πορεία τους γίνεται πιο ομαλή μεταξύ εποχών, πράγμα που δείχνει πιο σίγουρη και σταθερή μαθησιακή συμπεριφορά. Τα ευρήματα υποδεικνύουν ότι μετά το 24%, το μοντέλο περνά σε ένα πιο αξιόπιστο καθεστώς βελτιστοποίησης. Τα ποσοστά στην περιοχή 24–42% φαίνεται να έχον την καλύτερη απόδοση όσο αφορά την ομαλότητα και την τελική τιμή την νόρμας. 

Στα 48% και 54% ενώ η πορεία είναι πιο ομαλή από τα 6-18% σε απόλυτη τιμή η νόρμες αυξάνονται.


### 4) Χάσμα γενίκευσης

Για την ολοκλήρωση της ανάλυσης, υπολογίζω το χάσμα γενίκευσης (generalization gap) για κάθε ποσοστό. Το χάσμα γενίκευσης ορίζεται ως η διαφορά μεταξύ του σφάλματος εκπαίδευσης και του σφάλματος αξιολογησης μόνο στην τελευταία εποχή, με σκοπό να μετρηθεί η ικανότητα γενίκευσης όταν πλέον η εκπαίδευση έχει τελειώσει.

Μικρό χάσμα υποδηλώνει καλύτερη γενίκευση, ενώ μεγάλο αρνητικό χάσμα φανερώνει υπερπροσαρμογή. Και σεε αυτό το βήμα, υπολογίζω τον μέσο όρο του χάσματος γενίκευσης όλων των μετοχών και απεικονίζω την εξέλιξή του. Επειδή πήρα τον μέσο όρο, έχω βάλει και την τυπική απόκλιση για να φαίνεται σε τι διάστημα κυμαίνεται το χάσμα.


<img src="results/gengap.png" alt="Chart" width="700"/>


Το γράφημα του χάσματος γενίκευσης δείχνει ότι σε χαμηλά ποσοστά ( 6%, 12%), το μοντέλο παρουσιάζει μεγάλα αρνητικά χάσματα γενίκευσης, γεγονός που σημαίνει ότι αποδίδει σημαντικά χειρότερα στο σφάλμα αξιολόγησης σε σύγκριση με το σφάλμα εκπαίδευσης, επιβεβαιώνοντας όπως και στα προηγούμενα βήματα ότι σε αυτά τα ποσοστά γίνεται υπερπροσαρμογή.

Καθώς το ποσοστό αυξάνεται πέρα από το 18%, το χάσμα μειώνεται σταθερά και τελικά πλησιάζει το μηδέν, υποδηλώνοντας βελτιωμένη γενίκευση. Φαίνεται πως το χάσμα σταθεροποιείται κοντά στο μηδέν για ποσοστά 42%-54%.

Επιπλέον, η τυπική απόκλιση του χάσματος είναι μεγαλύτερη στα χαμηλότερα ποσοστά ακόμη και στο εύρος 24-36%. Συνολικά, η τάση αυτή επιβεβαιώνει ότι η αύξηση του ποσοστού εκπαίδευσης βελτιώνει τη γενίκευση και μειώνει τη διακύμανση απόδοσης του μοντέλου μεταξύ διαφορετικών μετοχών.

### 5) Συμπέρασμα - Το καλύτερο ποσοστό δεδομένων εκπαίδευσης

Σύμφωνα με τα αποτεέσματα όλων των αναλύσεων το **fewshot fraction = 42%** αναδεικνύεται ως το πιο αποδοτικό ποσοστό.

Σε αυτό το ποσοστό, τα σφάλματα αξιολόγησης παραμένουν σταθερά χαμηλά για τις περισσότερες μετοχές, δείχνοντας ισχυρή ικανότητα γενίκευσης. Οι νόρμες στο 42% έχουν σταθερή φθίνουσα πορεία σε κάθε εποχή, κάτι που δείχνει αποδοτική και σταθερή μάθηση, χωρίς ενδείξεις υπερπροσαρμογής.

Επιπλέον, το χάσμα γενίκευσης στο 42% είναι πολύ κοντά στο μηδέν και παρουσιάζει χαμηλή διακύμανση.
