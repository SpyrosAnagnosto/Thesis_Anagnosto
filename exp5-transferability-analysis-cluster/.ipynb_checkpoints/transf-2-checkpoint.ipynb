# Πείραμα 4 - Ανάλυση Μεταφερσιμότητας μέσω συσταδοποίησης - Εκπαίδευση σε μικτό σύνολο μετοχών

Είδαμε πως η εκπαίδευση στον δείκτη SPY βελτιώνει την επίδοση στο test set, αλλά από τα αποτελέσταμα στα παράθυρα γεγονότων φάνηκε πως το μοντέλο έχει δυσκολία να γενικεύσει σωστά ως προς την μεταβλητότητα των χρονοσειρών. Σε αυτό το πείραμα θα μεγαλόσουμε λίγο το σύνολο δεδομένων εκπαίδευσης με σκοπό να κάνουμε τις προβλέψεις πιο ανθεκτικές. Αλλά όπως είδαμε στο πείραμα 2 μεγαλώνοντας απλώς το ποσοστό δεδομένων εκπαιδευσης στον SPY θα οδηγήσει σε προβλήματα υπερπροσαρμογής. Οπότε πρέπει να πάρουμε δεδομένα από άλλες μετοχες. Για να το κάνουμε αυτό  στην μελέτη σφαλμάτων δημιουργούμε χάρτες συστάδων. Ωστόσο για τις κατευθυντικές μεντρικές οι χάρτες συστάδων δεν έβγαζαν αποτελέσματα με νόημα, πιθανών γιατί είναι πολύ μικές οι διαφορές μεταξύ των τιμών οπότε χρησιμοποιήσαμε γραφους μεταφερσιμότητας.

#### Ερευνητικά Ερωτήματα Πειράματος 4 – Ανάλυση Μεταφερσιμότητας μέσω συσταδοποίησης - Εκπαίδευση σε μικτό σύνολο μετοχών:
- Μπορούμε να κάνουμε καλύτερη ανάλυση για την ποιότητα της μεταφερσιμότητας μοτίβων μεταξύ των μετοχών;
  
- Μπορούμε να βρούμε ένα σύνολο μετοχών με όσο το δυνατό μικρότερο μέγεθος που θα βελτιώσει σημαντικά την απόδοση του προηγούμενου μοντέλου;


#### 2) Συσταδοποίηση βάση σφαλμάτων
Στο προηγούμενο βήμα μελετήσαμε την μεταφερσιμότητα των μοντέλων κοιτώντας τις τιμές των αποτελεσμάτων αυτές καθ' αυτές. Σε αυτό το βήμα παρατηρούμε τα αποτελέσματα πιο συγκεντρωτικά με σκοπό να καταλήξουμε σε μια πιο πλήρη εικόνα. 
Το πρώτο βήμα είναι, από τους θερμικούς χάρτες να κατασκευάσουμε χάρτες συστάδων (clustermaps). Ο χάρτης συστάδων παίρνει τον θερμικό χάρτη και ομαδοποιεί παρόμοιες γραμμές και στήλες. 

Όταν γραμμές ομαδοποιούνται, σημαίνει ότι τα μοντέλα που έχουν εκπαιδευτεί σε αυτές τις μετοχές κάνουν παρόμοια σφάλματα σε άλλες μετοχές. Αυτό δείχνει ότι τα μοντέλα μαθαίνουν από τα δεδομένα με τρόπο που γενικεύεται παρόμοια. Κάποιες φορές, η συσταδοποίηση ομαδοποιεί μετοχές που δεν φαίνεται να σχετίζονται ούτε από άποψη κλάδου, ούτε μεταβλητότητας. Αυτό μπορεί να υποδηλώνει υπόγειες ομοιότητες στη δυναμική της χρονοσειράς. Μοντέλα που βρίσκονται σε απομονωμένες συστάδες  τείνουν να συμπεριφέρονται πολύ διαφορετικά. Όταν παρατηρείται η ίδια συσταδοποίηση σε RMSE, MAE και MAPE, ενισχύεται η εμπιστοσύνη ότι οι σχέσεις είναι πραγματικές και όχι στατιστικός θόρυβος. Η συνέπεια αυτή σημαίνει ότι τα μοντέλα είναι παρόμοια όχι μόνο ως προς το συνολικό σφάλμα, αλλά και ως προς τη δομή και φύση των σφαλμάτων.

Αν δύο στήλες ανήκουν στην ίδια συστάδα, σημαίνει ότι τα περισσότερα μοντέλα κάνουν παρόμοια σφάλματα όταν προβλέπουν αυτές τις δύο μετοχές.
Έτσι, συσταδοποίηση στις στήλες δείχνει αν υπάρχει κάποιο συστηματικό μοτίβο σφάλματος για συγκεκριμένους στόχους.Ορισμένοι στόχοι μπορεί να δέχονται χαμηλά σφάλματα από όλα τα μοντέλα, και έτσι να συγκεντρώνονται σε συστάδες εύκολων στόχων. Άλλοι μπορεί να συσχετίζονται ως δύσκολοι στόχοι, ιδιαίτερα αν είναι ευμετάβλητοι. Αυτή η ομαδοποίηση βοηθά στο να αποφασίσουμε ποιες μετοχές χρειάζονται πιο σύνθετα ή προσαρμοσμένα μοντέλα. Επίσης, αν μια ολόκληρη συστάδα στόχων είναι συνεχώς δύσκολη στην πρόβλεψη, ίσως χρειάζεται να επανεξετάσουμε τις εισόδους του μοντέλου.

Όμως η SPY στο διάγραμμα του MAE είναι απομονωμένη δείχνοτας πως στην πρόβλεψή της είναι δύσκολος στόχος από την σκοπιά των τυπικών αποκλίσεων. Ως προς το MAPE η XOM συσταδοποιείται με τις TSLA και AMD, δείχνοντας ότι ποσοστιαία σε αυτές τις μετοχές υπάρχει δυσκολία στην πρόβλεψη. Από τα παραπάνω καταλήγουμε στο συμπέρασμα πως TSLA, SPY, AMD και ποσοστιαία XOM χρειάζονται πιο προσαρμοσμένα μοντέλα.

Και στα τρία διαγράμματα η συσταδοποίηση των γραμμών είναι η ίδια.  Μιά ενδιαφέρουσα παρατήρηση είναι ότι η JNJ δεν ανήκει στην ομάδα των SPY, XOM και PG. Αντιθέτως, η JNJ συσχετίζεται πρώτα με την TSLA, και στη συνέχεια αυτή η υπο-ομάδα συγχωνεύεται με τις AAPL και AMD. Αυτή η συσταδοποίηση δείχνει ότι τα μοντέλα που εκπαιδεύονται στην JNJ έχουν προφίλ γενίκευσης πιο κοντά σε ασταθείς μετοχές. Θα μπορόυσαμε να χρησιμοποιήσουμε την JNJ για την πρόβλεψη της TSLA χωρίς να θπάρχει κίνδυνος υπερπροσαρμογής. Οι AAPL και AMD σχηματίζουν την επόμενη κοντινή ομάδα με παρόμοιες επιδόσεις γενίκευσης, οπότε μοντέλα που εκπαιδεύονται σε AAPL μπορούν θα βρουν μοτίβα και για την πρόβλεψη της AMD . Αντίθετα, οι SPY, PG και XOM συγκεντρώνονται μαζί, γεγονός που επιβεβαιώνει πως τα μοντέλα που εκπαιδεύονται σε αυτές έχουν υψηλή μεταφερσιμότητα και γενική αξιοπιστία όταν εφαρμόζονται σε άλλες μετοχές.

**Συμπέρασμα:**

#### 3) Γράφοι μεταφερσιμότητας κατεύθυνσης
Για να μελετήσουμε την ποιότητα της μεταφερσιμότητας στην κατεύθυνση χρησιμοποιούμε γράφους μεταφερσιμότητας. Σε αυτούς του γράφους κάθε κόμβος αντιπροσοπεύει μια μετοχή που μπορεί να χρησιμοποιηθεί είτε ως σύνολο για εκπαίδευση, είτε ως στόχος για πρόβλεψη. Οπότε για κάθε δυάδα κόμβο υπάρχουν δύο ακμές, μία εισόδου και μία εξόδου. Αυτές τις ακμές τις ενώνουμε παίρνοντας έναν μέσο όρο. Στη συνέχεια τις χρωματίζουμε με βάση την τιμή αυτού του μέσου όρου. Πιο παχιές ή πιο σκούρες ακμές να σημαίνουν χειρότερη απόδοση. Σκοπός μας είναι πάλι να βρούμε αν μετοχές εκτός της SPY θα μπορέσουν να μας βοηθήσουν σε καλύτερες προβλέψεις.

<img src="results/graphs.png" alt="Chart"/>


Στο γράφημα του **Hit Rate**, η XOM παρουσιάζει ισχυρή ικανότητα μεταφοράς, καθώς έχει τις περισσότερες ανοιχτόχρωμες συνδέσεις. Για παράδειγμα, πετυχαίνει hit rate 0.496 στην TSLA, 0.492 στην SPY και 0.504 στην PG. Η SPY επίσης έχει ικανοποιητική μεταφορά, με hit rates όπως SPY → TSLA στο 0.492 και SPY → PG στο 0.490.

Στο **Recall** κυριαρχεί η SPY με SPY → AMD στο 0.562, AAPL → JNJ στο 0.557, SPY → AAPL στο 0.552. Επίσης έχουμε είναι η XOM → PG με τιμή 0.575 και AAPL → AMD στο 0.551. Αυτά δείχνουν ότι οι SPY, AAPL και XOM είναι ισχυρές πηγές μοντέλων που διατηρούν καλή απόδοση σε false negatives. 

Μεταξύ των ισχυρότερων ακμών στο **Precision** συγκαταλέγονται οι SPY → PG με τιμή precision 0.530, η υψηλότερη στο γράφημα, καθώς και οι SPY → TSLA (0.525), AMD → SPY (0.520), XOM → SPY (0.523) και XOM → JNJ (0.519). 

Στο **F1** η SPY έχει απόλυτη υπεροχή. Συγκεκριμένα, η μεταφορά από SPY προς PG επιτυγχάνει F1-score 0.550. Οι SPY → AMD (0.549) και SPY → TSLA (0.541).

**Συμπέρασμα:**
Από τα αποτελέσματα αυτά καταλήγουμε στο συμπέρασμα πως στο μοντέλο εκπαίδευσης με συνδυασμό πολλών μετοχών που καταλήξαμε παραπανω (SPY, JNJ, AAPL), μπορούμε να εντάξουμε και την XOM καθώς έχει καλή μεταφερσιμότητα όσο αφορά την κατεύθυνση.

### 3) Εκπαίδευση μικτού μοντέλου

Εκπαιδεύουμε το ΤΤΜ στο 42% των SPY, JNJ, AAPL, XOΜ, δηλαδή σε 6400 γραμμές δεδομένων.

Η απόδοση στο test set βελτιώνεται εμφανώς σε σχέση με την εκπαίδευση στην SPY, καθώς οι πραγματικές και οι προβλεπόμενες τιμές ευθυγραμμίζονται στενά σε όλο το χρονικό εύρος για το μεγαλύτερο μέρος των μετοχών. Οι πραγματικές και οι προβλεπόμενες τιμές ευθυγραμμίζονται στενά σε όλο το χρονικό εύρος για το μεγαλύτερο μέρος των μετοχών. Οι αποκλίσεις είναι μικρές και τείνουν να εμφανίζονται κυρίως σε περιόδους αιχμών ή έντονης μεταβλητότητας (ιδίως σε TSLA και AMD), χωρίς όμως να επηρεάζουν δραστικά την κατεύθυνση της πρόβλεψης.

Στις AAPL, SPY, XOM, PG, JNJ η πρόβλεψη ακολουθεί πιστά την πραγματική πορεία με ελάχιστες αστοχίες σε βραχυπρόθεσμες διακυμάνσεις.

Στις TSLA και AMD, αν και το μοντέλο αποτυπώνει τη γενική τάση, οι μεγάλες διακυμάνσεις τιμών οδηγούν σε στιγμιαία undershooting ή overshooting, κυρίως στις κορυφώσεις και τα τοπικά χαμηλά.

Η συνοχή μεταξύ μετοχών διαφορετικής μεταβλητότητας υποδηλώνει ότι το μικτό σύνολο εκπαίδευσης έχει συμβάλει στην ενίσχυση της γενίκευσης, προσφέροντας στο μοντέλο μοτίβα που είναι ανθεκτικά σε διαφορετικές δυναμικές αγοράς.

Στην Κατάρρευση & Ανάκαμψη (AAPL) και στην Κορύφωση Πετρελαϊκού Κύκλου (XOM), η πρόβλεψη ακολουθεί την κατεύθυνση με μικρές αποκλίσεις στο μέγεθος των κινήσεων.

Στο Drawdown Chop (SPY) και στη Χαμηλή Μεταβλητότητα (JNJ), οι αποκλίσεις είναι ελάχιστες, δείχνοντας ότι το μοντέλο έχει ενισχύσει την ικανότητα γενίκευσης σε διαφορετικά καθεστώτα αγοράς.

Στην Ηρεμία Ανεπηρέαστη από Μακροοικονομικούς Παράγοντες (PG), η πρόβλεψη είναι σχεδόν ταυτόσημη με την πραγματική τιμή.μ

#### Αποτελέσματα ανά καθεστώς μεταβλητότητας
Ο πίνακας συνοψίζει την απόδοση του μοντέλου ανά καθεστώς μεταβλητότητας, ομαδοποιόντας αποτελέσματα μετοχών παρόμοιας μεταβλητότητας:

* Στο καθεστώς **Υψηλής Μεταβλητότητας**
Το μοντέλο δίνει υψηλότερα σφάλματα (MAE 5.62, RMSE 8.12, MAPE 4.14%) λόγω των απότομων, απρόβλεπτων κινήσεων. Έχει μέτρια κατευθυντική εντοπίζει σωστά την κατεύθυνση περίπου στο 48% των περιπτώσεων, αλλά το εύρος των κινήσεων είναι δύσκολο να πιαστεί ακριβώς.

* Στο καθεστώς **Χαμηλής Μεταβλητότητας** το μοντέλο έχει τα χαμηλότερα σφάλματα όλων (MAE 1.43, RMSE 1.95, MAPE 1.08%), καθώς η αγορά κινείται ομαλά. Με υψηλό precision (0.587) έχει πολύ μικρό ποσοστό ψευδώς θετικών σημάτων.

* Στο καθεστώς **Αντιστροφής Τάσης** Tο μοντέλο έχει ενδιάμεσες τιμές σφάλματος (MAE 1.92, RMSE 2.56, MAPE 1.32%), αντανακλώντας μέτρια μεταβολή μετά από σημαντικές αλλαγές τάσης. Τα recall (0.535) και precision (0.520) είναι ισορροπημένα περίπου στις ίδιες τιμές.

* Στο καθεστώς **Πλάγιας/Ακατάστατης Κίνησης** το τα σφάλματα είναι μέτρια (MAE 3.89, RMSE 5.19, MAPE 1.83%). Το υψηλό recall (0.580) δείχνει ότι εντοπίζονται σχεδόν το 58% των πραγματικών ανοδικών κινήσεων, αλλά η διαφορά με το precision (0.526) δείχνει πως υπάρχουν ψευδώς θετικά σήματα.ην ισορροπημένη κατεύθυνση σε δύσκολα, ασταθή περιβάλλοντα.

#### Συγκρίσεις με το μοντέλο εκπαιδευμένο στην SPY

Αν συγκρίνουμε το μικτό μοντέλο με το μοντέλο που είχε εκπαιδευτεί μόνο στον SPY, προκύπτουν τα εξής βασικά συμπεράσματα:

- Το MAPE μειώνεται σε όλες τις μετοχές, με τη TSLA να έχει την πιο μεγάλη ποσοστιαία πτώση (−5.5 %), ακολουθούμενη από AMD (−2.5 %) και AAPL/XOM (−1.5 % έως −2 %). Μια παρατήρηση είναι πως όσο πιο ασταθής είναι η μετοχή, τόσο μεγαλύτερο το ποσοστιαίο όφελος από την συνθετική εκπαίδευση.

- Για την TSLA το RMSE πέφτει κατά περίπου 13 και το MAE κατά 10. Σημαντικότερο όφελος όλων, καθώς η τεχνολογική αυτή μετοχή έχει υψηλή βραχυπρόθεσμη μεταβλητότητα και ωφελείται περισσότερο από την ποικιλία σημάτων της εκπαίδευσης.

- Για τη SPY το RMSE και το MAE αυξάνονται με το συνθετικό μοντέλο άρα οι απόλυτες αποκλίσεις και τα μεγάλα σφάλματα της πρόβλεψης χειροτερεύουν. Θα μπορούσαμε να πούμε πως αυτό είναι αναμενόμενο καθώς ένα μοντέλο εκπαιδευμένο μόνο στην SPY είναι πιο εξειδικευμένο σε αυτή. Το MAPE όμως μειώνεται άρα, το ποσοστό λάθους σε σχέση με το μέγεθος της τιμής βελτιώνεται. Άρα, το μοντέλο όντως χάνει σε ακρίβεια τιμών αλλά προβλέπει Καλύτερο σε αναλογία με το πόσο ψηλά ή χαμηλά κινείται η SPY.
