# Πείραμα 7 - ARIMA
#### Διπλωματική εργασία - Αναγνωστόπουλος Σπύρος

### 1) Περιγραφή πειράματος
Το **ARIMA** είναι ένα από τα πιο γνωστά στατιστικά μοντέλα για πρόβλεψη χρονοσειρών. Η λέξη **ARIMA** είναι ακρωνύμιο που σημαίνει:
- **AR** – *AutoRegressive* (Αυτοπαλίνδρομο):
   Το μοντέλο προβλέπει την τρέχουσα τιμή με βάση προηγούμενες τιμές της ίδιας χρονοσειράς, π.χ. Αν η τιμή σήμερα μοιάζει με μέσο όρο των τιμών των τελευταίων 3 ημερών, τότε έχουμε AR(3).
- **I** – *Integrated* (Ολοκληρωμένο / Διαφορές):
   Αν η χρονοσειρά έχει τάση ή δεν είναι στατική (στατιστικά σταθερή στο χρόνο), παίρνουμε διαφορές για να αφαιρέσουμε την τάση και να κάνουμε τα δεδομένα σταθερά , π.χ. αν κάθε μέρα οι τιμές αυξάνονται κατά 1, παίρνοντας τη διαφορά (σήμερα – χθες) παίρνουμε σταθερή σειρά.
- **MA** – *Moving Average* (Κινητός Μέσος): το μοντέλο χρησιμοποιεί τα λάθη πρόβλεψης από προηγούμενες μέρες για να βελτιώσει την πρόβλεψη, π.χ. αν χθες κάναμε πρόβλεψη και πέσαμε έξω κατά +2, ίσως σήμερα πρέπει να διορθώσουμε προς τα κάτω.


Το μοντέλο γράφεται σαν **ARIMA(p, d, q)**:
* **p** → αριθμός καθυστερήσεων στο AR μέρος (πόσες προηγούμενες τιμές λαμβάνουμε υπόψη).
* **d** → αριθμός διαφορών (για να κάνουμε τη σειρά στάσιμη).
* **q** → αριθμός καθυστερήσεων στο MA μέρος (πόσα προηγούμενα σφάλματα λάμβάνουμε υπόψη).




#### Ερευνητικά Ερωτήματα Πειράματος 7 – ARIMA
- Ποιοι είναι οι καλύτεροι όροι p,d,q;
- Αφού βρούμε τους καλύτερους όρους πόσο καλά εφαρμόζεται το ARIMA στα δεδομένα μας;
- Σε αντίθεση με το ΤΤΜ το ARIMA πρέπει να εκπαιδευτεί σε κάθε χρονοσειρά ξεχωριστά, άρα χρειάζεται συνολικά περισσότερα δεδομένα. Έχουμε με περισσότερα δεδομένα αρκετά καλύτερη επίδοση, σε σχέση με το καλύτερο ΤΤΜ;
- Ποια είναι η οικονομική αξία των προβλέψεων

### 2)Έλεγχος στασιμότητας με το ADF test

Πριν εφαρμόσουμε το μοντέλο ARIMA, πρέπει να βεβαιωθούμε ότι η χρονοσειρά είναι στασιμή. Μια στασιμή χρονοσειρά έχει σταθερό μέσο όρο, σταθερή διακύμανση και η συσχέτιση μεταξύ παρατηρήσεων εξαρτάται μόνο από την χρονική τους απόσταση και όχι από το πότε έγιναν οι μετρήσεις.
Αν η σειρά δεν είναι στασιμή, οι υποθέσεις του ARIMA μπορεί να παραβιάζονται και τα αποτελέσματα να είναι παραπλανητικά.

Για τον έλεγχο στασιμότητας χρησιμοποιούμε το **ADF test (Augmented Dickey–Fuller test)**.Το τεστ αυτό βασίζεται στη λογική των στατιστικών υποθέσεων:

* **H₀ (μηδενική υπόθεση)**: Η σειρά δεν είναι στασιμή.
* **H₁ (εναλλακτική υπόθεση)**: Η σειρά είναι στασιμή.

Το ADF test μας δίνει δύο τρόπους να εγκρίνουμε υποθέσεις. 
1) **p-value:** αν το p-value < 0.05, απορρίπτουμε την H₀, δηλαδή η σειρά θεωρείται στασιμή. Αν το p-value ≥ 0.05, εγκρίνουμε την H₀ και η σειρά δεν θεωρείται στασιμή.

2) **ADF statistic & Critical values:** Το ADF statistic είναι η υπολογισμένη τιμή του στατιστικού ελέγχου. Τα critical values (1%, 5%, 10%) είναι τα κατώφλια για τρία διαφορετικά επίπεδα σημαντικότητας. Αν το ADF statistic είναι μικρότερο από ένα critical value, απορρίπτουμε την H₀ στο αντίστοιχο επίπεδο σημαντικότητας.Για παράδειγμα, αν το ADF statistic είναι μικρότερο από το critical value 5%, τότε η σειρά θεωρείται στασιμή στο επίπεδο σημαντικότητας 5%.:
     * 1% → πολύ αυστηρό κριτήριο (σφάλμα τύπου Ι ≤ 1%)
     * 5% → το πιο συνηθισμένο κριτήριο
     * 10% → πιο χαλαρό κριτήριο


Αν η σειρά δεν είναι στασιμή, εφαρμόζουμε διαφορική μετατροπή. Αντικαθιστούμε κάθε τιμή με τη διαφορά της από την προηγούμενη τιμή. Για παράδειγμα, αν έχουμε **τιμές close [100, 102, 105]**, οι διαφορές είναι **[102−100 = 2, 105−102 = 3]**. Αυτό αφαιρεί την τάση και κάνει τη σειρά πιο «σταθερή» στο χρόνο. Μερικές φορές χρειάζονται πολλές επαναλήψεις αυτής της διαδικασίας, αλλά στόχος είναι να βρούμε **το μικρότερο d** που κάνει τη σειρά στασιμή.

### 3) Διαγράμματα ACF, PACF

Αφού βεβαιωθήκαμε ότι η σειρά μας είναι στάσιμη για d=1, το επόμενο βήμα είναι να κοιτάξουμε αν υπάρχουν μοτίβα συσχέτισης στα δεδομένα.
Αν η σειρά παρουσιάζει συσχέτιση με προηγούμενες τιμές της, μπορούμε να χρησιμοποιήσουμε αυτή την πληροφορία στο ARIMA μέσω των παραμέτρων 
p(AR) και q(MA).

Για να το δούμε αυτό, χρησιμοποιούμε δύο διαγράμματα:

1) **ACF (Autocorrelation Function):** Μετράει πόσο μοιάζει η σημερινή τιμή της χρονοσειράς με τιμές που απέχουν 1 μέρα, 2 μέρες, 3 μέρες κ.ο.κ. (lags). Αν η μπάρα είναι μεγάλη για μικρά lags (εκτός του 0 που είναι η σημερινή τιμή), σημαίνει ότι η σημερινή τιμή εξαρτάται έντονα από τις πιο πρόσφατες τιμές και πιθανότατα θα χρειαστούμε όρους ΜΑ. Στα μοντέλα MA, η τρέχουσα τιμή επηρεάζεται από τυχαία σοκ των προηγούμενων ημερών. Αν το ACF δείχνει σημαντική συσχέτιση για λίγα πρώτα lags και μετά πέφτει, αυτό είναι τυπικό μοτίβο MA.

2) **PACF (Partial Autocorrelation Function):** Μετράει την καθαρή συσχέτιση ενός συγκεκριμένου lag με το παρόν, αφαιρώντας την έμμεση επίδραση των μικρότερων lags. Για παράδειγμα, αν θέλουμε να δούμε τη σχέση της σημερινής τιμής με την τιμή πριν 2 μέρες, το PACF θα αφαιρέσει την επίδραση που περνάει μέσω της τιμής πριν 1 μέρα. Αν υπάρχει ισχυρή καθαρή συσχέτιση για λίγα πρώτα lags, αυτό υποδηλώνει ότι οι τιμές εξαρτώνται άμεσα από προηγούμενες παρατηρήσεις και χρειαζόμαστε τιμές AR.

Στα διαγράμματα, επίσης βλέπουμε μπλε οριζόντιες γραμμές. Αυτές είναι τα όρια εμπιστοσύνης ( ±2/√N, όπου N είναι ο αριθμός των παρατηρήσεων). Αν η κορυφή μιας μπάρας μένει μέσα στα όρια, τότε η συσχέτιση που δείχνει είναι στατιστικά ασήμαντη. Αν η κορυφή μιας μπάρας βγαίνει έξω από τα όρια, η συσχέτιση θεωρείται στατιστικά σημαντική, δηλαδή ρίνσι πιθανό να υπάρχει κάποια πραγματική σχέση.

#### **Αποτελέσματα**
**ACF**: Όλες οι ράβδοι (εκτός του lag 0) κινούνται πολύ κοντά στο μηδέν και εντός της ζώνης εμπιστοσύνης. Δεν υπάρχει καθαρό κόψιμο ούτε αργή φθορά.

**PACF**: Αντίστοιχα, δεν εμφανίζονται σημαντικές αιχμές σε μικρά lags και οι ράβδοι είναι μικρές και εντός ζώνης.

Το ζεύγος ACF–PACF υποδηλώνει ότι, αφού αφαιρέσαμε την τάση με d=1, δεν μένει ουσιαστική γραμμική εξάρτηση που να δικαιολογεί AR ή MA όρους. Με άλλα λόγια, μια παραγοντικά λιτή προδιαγραφή όπως ARIMA(0,1,0) είναι υποψήφιο μοντέλο.

Αυτό στην πράξη σημαίνει πως ξεκινάμε με ARIMA(0,1,0) ως βάση και μετά θα χρειαστεί να ελέγξουμε μόνο πολύ μικρές προσθήκες (π.χ. AR(1) ή MA(1)) αν βελτιώνουν ουσιαστικά τις μετρικές. 

#### **Γενικό συμπέρασμα για όλες τις μετοχές**

Κάνουμε την ίδια διαδικασία για όλες τις μετοχές *(φάκελος results)*. Από την ανάλυση ΑCF, PACF βλέπουμε σε όλες τις μετοχές ένα κοινό μοτίβο. Οι ράβδοι (εκτός του lag 0) παραμένουν πολύ κοντά στο μηδέν και μέσα στα όρια εμπιστοσύνης. Δεν εμφανίζονται σημαντικές αιχμές ούτε σε χαμηλά lags. Δεν υπάρχει ένδειξη καθαρής, σταθερής αποσύνθεσης στο ACF, ούτε απότομης αποκοπής στο PACF. Αυτό υποδηλώνει πως η τάξη d=1 ήταν αρκετή για να αφαιρέσει την αυτοσυσχέτιση και η σειρά είναι πλέον στάσιμη. Δεν φαίνεται να απαιτούνται ισχυροί όροι AR ή MA. Ένα απλό μοντέλο όπως ARIMA(0,1,0) ή με πολύ μικρές προσθήκες (π.χ. AR(1) ή MA(1)) είναι πιθανώς επαρκές.

### 4) Grid search
Για να σιγουρευτούμε πως δεν χρειάζονται προσθήκες όρων ΑΡ,ΜΑ κάνουμε ένα grid serach. Το grid search είναι μια δοκιμή όλων των συνδυασμών για μικρά p,q του ARIMA, κρατώντας σταθερό το d που έχουμε βρει από το τεστ στασιμότητας , ώστε να βρούμε την πιο αποδοτική παραμετροποίηση για κάθε χρονοσειρά. Σκοπός αυτού του βήματος δεν είναι να κάνουμε προβλέψεις, αλλά να σιγουρευτούμε πως οι όροι AR,MA δεν παίζουν σημαντικό ρόλο.

Για μετοχή και για κάθε συνδυασμό (p,d,q):
1. Μοντελοποιούμε τη χρονοσειρά με τον συνδυασμό.
2. Κάνουμε προβλέψεις σε ένα μικρό test set.
3. Μετράμε την απόδοση με μετρικές.

#### **Αποτελέσματα**

Από τα αποτελέσματα του grid search φαίνεται ότι οι όροι AR και MA δεν βελτιώνουν σημαντικά την απόδοση στις περισσότερες περιπτώσεις.

Αυτό συμφωνεί και με την εικόνα από τα διαγράμματα ACF/PACF, όπου δεν εμφανίζονται έντονες συσχετίσεις σε μικρά lags που θα δικαιολογούσαν πιο πολύπλοκους όρους.

Με άλλα λόγια, η βασική πληροφορία για την πρόβλεψη φαίνεται να προέρχεται κυρίως από το τμήμα διαφοροποίησης (d), και όχι από προσθήκες AR ή MA υψηλής τάξης.

### 5) Βασική εκτίμηση και έλεγχος υπολοίπων
Αφού βρήκαμε ότι η σειρά γίνεται στάσιμη με μία διαφορική μετατροπή (d = 1), το επόμενο βήμα είναι να μοντελοποιήσουμε με ένα βασικό μοντέλο ARIMA(0, 1, 0) σε όλο το σύνολο δεδομένων. Σκοπός αυτού του βήματος δεν είναι να κάνουμε προβλέψεις, αλλά να κάνουμε διάγνωση σφαλμάτων. Αν αυτό το απλό μοντέλο περιγράφει επαρκώς τα δεδομένα, τότε η κατασκευή πιο σύνθετων μοντέλων δεν είναι απαραίτητη.

Το αποτέλεσμα της εκτίμησης δίνεται σε έναν στατιστικό πίνακα που περιλαμβάνει:
- Τιμές συντελεστών (π.χ. intercept, διακύμανση υπολοίπων)
- Ljung–Box: Ελέγχει αν τα υπόλοιπα έχουν αυτοσυσχέτιση (δηλαδή αν υπάρχει κάποια προβλέψιμη δομή που το μοντέλο δεν έχει πιάσει). Aν p-value > 0,05, τα υπόλοιπα είναι ασυσχέτιστα.
- Jarque–Bera: Ελέγχει αν η κατανομή των υπολοίπων είναι κανονική (συμμετρική, χωρίς βαριές ουρές). Αν p-value > 0,05 τότε είναι περίπου κανονική.
- Heteroskedasticity: Ελέγχει αν η διακύμανση των υπολοίπων είναι σταθερή στον χρόνο. Αν p-value > 0,05 υπάρχει σταθερή μεταβλητότητα.
- Δείκτες προσαρμογής (AIC, BIC).
- Διαστήματα εμπιστοσύνης για τους συντελεστές.

  
Όταν εφαρμόζουμε ένα μοντέλο ARIMA, τα υπόλοιπα είναι οι διαφορές ανάμεσα στις πραγματικές τιμές και στις τιμές που προβλέπει το μοντέλο. Αν το μοντέλο είναι σωστά προσαρμοσμένο τα υπόλοιπα πρέπει να μοιάζουν με λευκό θόρυβο (white noise), δηλαδή να είναι ασυσχέτιστα, με σταθερή διακύμανση και μηδενική μέση τιμή.Η εντολή plot_diagnostics() της βιβλιοθήκης statsmodels δημιουργεί 4 τυπικά διαγράμματα:

- **Standardized residuals over time**: Γραφική παράσταση των υπολοίπων στον χρόνο η οποία μας δείχνει αν υπάρχουν συστηματικά μοτίβα ή αλλαγές στη διακύμανση. Θέλουμε να δούμε τυχαία κατανομή γύρω από το 0 χωρίς ορατή δομή.

- **Histogram & KDE plot**: Ιστόγραμμα των υπολοίπων, με υπερτιθέμενη εκτίμηση πυκνότητας (KDE) και την καμπύλη κανονικής κατανομής N(0,1). Μας δείχνει αν τα υπόλοιπα ακολουθούν κατά προσέγγιση κανονική κατανομή.

- **Normal Q–Q plot**: Συγκρίνει την κατανομή των υπολοίπων με την ιδανική κανονική κατανομή. Αν τα σημεία βρίσκονται πάνω στη διαγώνιο, η κατανομή είναι κοντά στην κανονική. Αποκλίσεις στις άκρες υποδεικνύουν βαριές ουρές.

- **Correlogram (ACF των υπολοίπων)**: Δείχνει τις αυτοσυσχετίσεις των υπολοίπων για διάφορα lags. Θέλουμε οι μπάρες να βρίσκονται μέσα στα μπλε όρια εμπιστοσύνης ως ένδειξη ότι δεν μένει σημαντική αυτοσυσχέτιση.

#### **Αποτελέσματα**

- **Intercept = 0.0633**: Κατά μέσο όρο, η σειρά τείνει να αυξάνεται κατά περίπου 0.0633 μονάδες ανά χρονικό βήμα μετά τη διαφορική μετατροπή.

- **Std err (0.025)**: Εκτίμηση της αβεβαιότητας του intercept. Ο λόγος (z = 2.534) και το p-value = 0.011 δείχνο ότι το intercept είναι στατιστικά σημαντικό.

- **Sigma² = 2.4206**
Εκτίμηση της διακύμανσης των υπολοίπων. Εδώ σημαίνει ότι οι διαφορές που μένουν ανεξήγητες από το μοντέλο έχουν τυπική απόκλιση περίπου √2.4206 ≈ 1.55.

- **Ljung–Box Q (1.59, p = 0.21)**: Ελέγχει αν μένει σημαντική αυτοσυσχέτιση στα υπόλοιπα. P-value > 0.05 σημαίνει πως δεν απορρίπτουμε την υπόθεση ότι τα υπόλοιπα είναι ασυσχέτιστα.

- **Heteroskedasticity (111.91, p < 0.001)**: Ελέγχει αν η διακύμανση των υπολοίπων είναι σταθερή στον χρόνο. Εδώ το p-value είναι πολύ μικρό πράγμα που δείχνει πως η υπάρχει ετεροσκεδαστικότητα, άρα η μεταβλητότητα αλλάζει στον χρόνο.

- **Jarque–Bera (19478.90, p < 0.001)**: Ελέγχει αν τα υπόλοιπα ακολουθούν κανονική κατανομή. Εδώ το πολύ μικρό p-value δείχνει ότι τα υπόλοιπα δεν είναι κανονικά (έχουν ασυμμετρία και/ή βαριές ουρές).

- **Τυποποιημένα υπόλοιπα στον χρόνο**<br>
Τα υπόλοιπα ταλαντεύονται γύρω από το μηδέν χωρίς εμφανή περιοδικά μοτίβα, γεγονός που ενισχύει την ένδειξη απουσίας συστηματικής αυτοσυσχέτισης. Ωστόσο, υπάρχει σαφής χρονικά μεταβαλλόμενη μεταβλητότητα μετά το 2020. Αυτό συμφωνεί με το αποτέλεσμα του ελέγχου ετεροσκεδαστικότητας, το οποίο επιβεβαιώνει τη μη σταθερότητα της διακύμανσης στον χρόνο.

- **Ιστόγραμμα με εκτιμώμενη πυκνότητα**<br>
Το ιστόγραμμα και η εκτίμηση πυκνότητας με πυρήνα (πορτοκαλί καμπύλη) παρουσιάζουν πιο έντονη κορυφή και βαρύτερες ουρές σε σύγκριση με την καμπύλη κανονικής κατανομής (πράσινη γραμμή). Αυτή η οπτική ένδειξη υποστηρίζει τα αποτελέσματα του ελέγχου Jarque–Bera, δείχνοντας ισχυρές αποκλίσεις από την κανονικότητα λόγω υπερβάλλουσας κύρτωσης (βαριές ουρές) και ασυμμετρίας.

- **Διάγραμμα Normal Q–Q**<br>
Σε μια τέλεια κανονική κατανομή, τα σημεία θα έπρεπε να βρίσκονται πάνω στην κόκκινη γραμμή των 45°. Στην παρούσα περίπτωση η κεντρική περιοχή ακολουθεί περίπου τη γραμμή, αλλά οι ουρές αποκλίνουν. Η δεξιά ουρά βρίσκεται πάνω από τη γραμμή, ενώ η αριστερή ουρά κάτω από αυτήν, υποδεικνύοντας πως οι ακραίες τιμές υπολοίπων εμφανίζονται συχνότερα απ’ ό,τι θα αναμενόταν υπό κανονικότητα. Αυτό συμφωνεί με την υψηλή κύρτωση (≈ 13,96) που αναφέρεται στον συνοπτικό πίνακα.

- **Διορθωμένο διάγραμμα αυτοσυσχέτισης**<br>
Όλες οι μπάρες αυτοσυσχέτισης βρίσκονται εντός των ορίων εμπιστοσύνης 95% (σκιασμένη περιοχή), εκτός από την υστέρηση 1, η οποία είναι μικρή και στατιστικά ασήμαντη. Αυτό επιβεβαιώνει το αποτέλεσμα του ελέγχου Ljung–Box ότι δεν υπάρχει ένδειξη σειριακής συσχέτισης στα υπόλοιπα, πράγμα που σημαίνει ότι το ARIMA(0,1,0) έχει αφαιρέσει επιτυχώς τη γραμμική εξάρτηση από τη σειρά.


#### **Γενικό συμπέρασμα για όλες τις μετοχές**

Η παραπάνω διαδικασία πρέπει να γίνει για όλες τις μετοχές *(φάκελος reuslts στο exp7)*. Κοιτώντας συνολικά τα διαγράμματα υπολοίπων για TSLA, XOM, SPY, JNJ, AMD και PG, βλέπουμε μια επαναλαμβανόμενη εικόνα. Oι ARIMA(0,1,0) προσεγγίσεις που εφαρμόστηκαν περιγράφουν σε μεγάλο βαθμό τη χρονική εξάρτηση των δεδομένων, για τα υπόλοιπα υπάρχουν όμως ενδείξεις μη σταθερής διακύμανσης και αποκλίσεις από την κανονικότητα.

Στα γραφήματα **Standardized residuals** παρατηρούμε ότι, σχεδόν σε όλες τις σειρές, η διακύμανση τν υπολοίπων δεν είναι απολύτως σταθερή με τον χρόνο, υπάρχουν περίοδοι αυξημένης ή μειωμένης μεταβλητότητας.
Το **Histogram & KDE plot** δείχνει ότι οι κατανομές των υπολοίπων είναι πιο απότομες και με πιο βαριές ουρές σε σχέση με την κανονική κατανομή,πράγμα που επιβεβαιώνεται και από τα **Q-Q plots**.
Το **correlogram** είναι γενικά καλό για όλες τι μετοχές (οι μπάρες είναι μέσα στα όρια), κάτι που σημαίνει ότι η αυτοσυσχέτιση στα κατάλοιπα έχει σχεδόν εξαλειφθεί, άρα το μοντέλο έχει καταφέρει να απορροφήσει τη βασική δομή εξάρτησης της σειράς.
Για καλύτερη μοντελοποίηση, ειδικά όσον αφορά την αβεβαιότητα και τις προβλέψεις διαστημάτων εμπιστοσύνης, πιθανόν να χρειάζονται πιο εξελιγμένες τεχνικές, όπως GARCH ή μοντέλα που λαμβάνουν υπόψη την ετεροσκεδαστικότητα και τις βαριές ουρές.

### 6) Σχόλιο για τα αποτελέσματα

Η υπεροχή του απλού μοντέλου ARIMA(0,1,0) έναντι πιο σύνθετων προδιαγραφών με όρους AR ή MA στα δεδομένα τιμών μετοχών μπορεί να αποδοθεί σε δομικά χαρακτηριστικά των χρηματιστηριακών χρονοσειρών.

Πρώτον, οι τιμές μετοχών τείνουν να ακολουθούν μια διαδικασία τύπου random walk, όπου η καλύτερη πρόβλεψη της μελλοντικής τιμής είναι η τρέχουσα τιμή συν ένα μη συστηματικό σφάλμα. Στατιστικά, η συμπεριφορά αυτή οδηγεί σε έντονη μη στασιμότητα, η οποία όμως αντιμετωπίζεται επαρκώς με μία διαφορική μετατροπή (d = 1). Μετά τη διαφοροποίηση, η σειρά προσεγγίζει στάσιμη διαδικασία με ιδιότητες παρόμοιες με λευκό θόρυβο, περιορίζοντας την ωφέλεια πρόσθετων όρων AR ή MA.

Δεύτερον, οι χρηματιστηριακές τιμές επηρεάζονται σε μεγάλο βαθμό από εξωγενείς και συχνά απρόβλεπτους παράγοντες, όπως μακροοικονομικές ανακοινώσεις, εταιρικά αποτελέσματα ή γεωπολιτικά γεγονότα. Η φύση αυτών των γεγονότων περιορίζει την προβλεψιμότητα που μπορεί να αντληθεί αποκλειστικά από την εσωτερική χρονική δομή της σειράς.

Τέλος, η όποια υφιστάμενη εξάρτηση εμφανίζεται συνήθως ως πολύ βραχυπρόθεσμο φαινόμενο, το οποίο αποτυπώνεται σε χαμηλής τάξης όρους AR(1) ή MA(1) και συχνά δεν επαρκεί για να επιφέρει ουσιαστική βελτίωση σε μετρικές πρόβλεψης όπως RMSE ή MAPE.

Η εμπειρική διαπίστωση ότι το ARIMA(0,1,0) επιτυγχάνει απόδοση συγκρίσιμη ή και ανώτερη από πιο πολύπλοκες παραμετροποιήσεις είναι αναμενόμενη υπό το πρίσμα των παραπάνω ιδιοτήτων. Το κύριο όφελος στην πρόβλεψη προέρχεται από τη διαφοροποίηση (d = 1) που καθιστά τη σειρά στάσιμη, ενώ η περαιτέρω πολυπλοκότητα δεν συνεπάγεται αναλογική βελτίωση της προβλεπτικής ικανότητας.

Όπως είπαμε παραπάνω, για καλύτερες προβλέψεις θα μπορούσαν να χρησιμοποιηθούν πιο συνθετα μοντέλα (GARCH). Σε αυτήν την εργασία θα χρησιμοποιήουμε το ARIMA(0,1,0) για όλες τις μετοχές. 

## Εκπαίδευση και αξιολόγηση

### 7) Οπτική αξιολόγηση του τελικού μοντέλου

Παρατηρήσεις από τα διαγράμματα (ARIMA(0,1,0) στο test set)

Στα σχήματα φαίνεται η τυπική «συμπεριφορά αδράνειας» του ARIMA(0,1,0): η πρόβλεψη ακολουθεί ομαλά την τελευταία τιμή και γι’ αυτό τείνει να υποεκτιμά το εύρος μεγάλων κινήσεων και να καθυστερεί στα γυρίσματα.

* **AAPL**: το μοντέλο ευθυγραμμίζεται καλά με την ανοδική πορεία, αλλά σε ράλι τύπου 2023–24 υποεκτιμά τα υψηλά (συστηματικό αρνητικό bias στις κορυφές). Η χρονική υστέρηση είναι μικρή αλλά ορατή στις απότομες διορθώσεις.
* **TSLA**: χαρακτηριστική αδυναμία. Στην εκρηκτική άνοδο 2020–21 και στις έντονες ταλαντώσεις του 2022 η πρόβλεψη «στρογγυλεύει» την κίνηση, υστερεί σε φάση και υποεκτιμά το μέγεθος τόσο στις κορυφές όσο και στους πυθμένες.
* **XOM**: μετά το 2021 η ανοδική τάση του πετρελαίου αποτυπώνεται ικανοποιητικά. Παρ’ όλα αυτά, στα τοπικά υψηλά του 2022–23 υπάρχει υστέρηση και συγκρατημένη εκτίμηση εύρους.
* **SPY**: η ευθυγράμμιση είναι γενικά καλή, με σαφές smoothing. Στη bear φάση του 2022 η υπόδειξη είναι σωστή κατεύθυνση αλλά μικρότερο βάθος πτώσης και πιο αργή ανάκαμψη από την πραγματική.
* **JNJ**: από τις καλύτερες περιπτώσεις. Η πρόβλεψη ακολουθεί στενά την πραγματική τιμή σε όλο το παράθυρο, με μικρές μόνο χρονικές καθυστερήσεις σε gap-like κινήσεις (π.χ. στο σοκ του 2020).
* **AMD**: όπως και η TSLA, παρουσιάζει έντονο underfitting. Οι μεγάλες αιχμές 2021–22 δεν αποδίδονται επαρκώς και οι αναστροφές έρχονται με καθυστέρηση.
* **PG**: πολύ κοντά στη γραμμή αναφοράς σε όλο το διάστημα, με μικρή υποεκτίμηση στα πρόσφατα υψηλά. Το χαμηλό beta της μετοχής ταιριάζει στο υπόδειγμα.

Σε συνολική αποτίμηση, τα γραφήματα επιβεβαιώνουν όσα είδαμε στους πίνακες: το ARIMA(0,1,0) **αποδίδει πειστικά σε χαμηλή μεταβλητότητα** (PG, JNJ και σε μεγάλο βαθμό XOM), παραμένει **επαρκές** σε πλαγιοτάσικα διαστήματα (SPY, AAPL), ενώ **υστερεί** εμφανώς σε περιβάλλοντα **υψηλής μεταβλητότητας και απότομων αναστροφών** (TSLA, AMD), όπου η εξομάλυνση και η χρονική υστέρηση οδηγούν σε υποεκτίμηση του εύρους και καθυστερημένες στροφές.

### 13) Οπτικοποίηση τελικής βελτιστοποίησης σε σχέση με το τελικό μοντέλο ΤΤΜ

**1) Θερμικοί χάρτες σφαλμάτων**<br>
Το heatmap απεικονίζει τη διαφορά σφάλματος μεταξύ ARIMA και τελικού μοντέλου για τις τρεις μετρικές πλάτους (MAE, RMSE, MAPE). Αρνητικές τιμές σημαίνουν ότι το ARIMA έχει μικρότερο σφάλμα από το τελικό μοντέλο. όσο πιο σκούρα/αρνητική η τιμή, τόσο μεγαλύτερο το πλεονέκτημα του ARIMA. Το πιο έντονο πλεονέκτημα του ARIMA εμφανίζεται στην TSLA ( MAE −1.79, RMSE −2.06, MAPE −1.14), υποδεικνύοντας σαφώς μικρότερες αποκλίσεις επιπέδου. Υπεροχή παρατηρείται και σε SPY (MAE −1.00, RMSE −1.19), καθώς και σε AMD (MAE −0.59, RMSE −0.70, MAPE −0.61). Σε JNJ και PG τα κενά μικραίνουν (MAE −0.25 και −0.21 αντίστοιχα), υποδηλώνοντας σχεδόν ισοδύναμη ακρίβεια επιπέδου. Συνολικά, ο χάρτης καταδεικνύει ότι το ARIMA τείνει να παρακολουθεί πιο στενά το επίπεδο της τιμής (μικρότερα MAE/RMSE/MAPE), με το πλεονέκτημα να μεγενθύνεται σε μετοχές υψηλής μεταβλητότητας. 

**2) Θερμικοί χάρτες κατευθυνσης**<br>
Το δεύτερο heatmap δείχνει τη μεταβολή των κατευθυντικών μετρικών όταν περνάμε από το ARIMA στο τελικό μοντέλο. Θετικές τιμές σημαίνουν ότι το τελικό μοντέλο υπερέχει, αρνητικές ότι υπερέχει το ARIMA. Ο πιο σταθερός κερδισμένος άξονας για το τελικό μοντέλο είναι το recall όπου εμφανίζει αισθητή βελτίωση σε AAPL (+0.050), SPY (+0.051) και AMD (+0.045) και μικρότερη σε JNJ/PG. Αυτό σημαίνει ότι το τελικό μοντέλο συλλαμβάνει περισσότερες από τις πραγματικές ανοδικές/καθοδικές κινήσεις, μειώνοντας τις ψευδώς αρνητικές τιμές. Το F1 βελτιώνεται ιδιαίτερα σε SPY (+0.023) και AMD (+0.022). Αντίθετα, το hit\_rate είναι οριακά χαμηλότερο για το τελικό μοντέλο στις περισσότερες μετοχές (−0.014 σε AAPL, −0.011 σε SPY, −0.025 σε PG), δείχνοντας ότι το ARIMA δίνει λίγο περισσότερα σωστά σήματα συνολικά, αλλά με χειρότερη ανάκληση στα πιο δύσκολα επεισόδια. Οι μεταβολές στο precision είναι μικρές και ανάμεικτες (≈0 έως −0.01), άρα δεν αλλάζει δραματικά η αναλογία σε ψευδώς θετικές προβλέψεις.

**3) Wilcoxon**<br> 
Στο επίπεδο σημαντικότητας 5% απορρίπτουμε την ισότητα για MAE, RMSE, MAPE, Hit-rate και Recall και άρα υπάρχει στατιστικά σημαντική διαφορά ανάμεσα στα δύο μοντέλα σε αυτές τις μετρικές. Για Precision και F1 η διαφορά δεν είναι στατιστικά σημαντική (p>0.05).

**Συμπέρασμα**<br>
Τα διαγράμματα επιβεβαιώνουν ένα σαφές μοτίβο. Το ARIMA υπερτερεί στις μετρικές πλάτους, ενώ το τελικό μοντέλο υπερέχει σε κατευθυντική ευαισθησία, αυξάνοντας το recall και το F1. Αυτό είναι συνεπές με την ανάλυση μεταβλητότητας στο ότι τα σφάλματα του ARIMA διογκώνονται περισσότερο όταν αυξάνει η μεταβλητότητα, ενώ το τελικό μοντέλο πιάνει καλύτερα την κατεύθυνση σε περιόδους πιο δύσκολης δυναμικής.

### 16)  Αποτελέσματα στρατιγικής - ποσοστιαία διαγράμματα

Ξεκινάμε από 0% κεφάλαιο και σε κάθε βήμα παίρνουμε θέση long όταν το μοντέλο προβλέπει άνοδο και θέση short όταν προβλέπει πτώση. Η καμπύλη κάθε μετοχής δεν είναι η τιμή της, αλλά το αποτέλεσμα του συγκεκριμένου κανόνα πάνω στη μετοχή. Αν η καμπύλη ανεβαίνει, σημαίνει ότι οι προβλέψεις έπιαναν σωστά το πρόσημο των κινήσεων εκείνη την περίοδο, αν κατεβαίνει, οι προβλέψεις ήταν, κατά μέσο όρο, αντίθετες από την πραγματική κίνηση. Οι απότομες βουτιές είναι drawdowns, ενώ ο τελικός ύψος της καμπύλης είναι η συνολική σωρευτική απόδοση.

Στο διάγραμμα (Strategy Cumulative Returns – ARIMA)  υπάρχουν σημαντικές διαφοροποιήσεις σε σχέση με το τελικό μοντέλο. Το ARIMA οδηγεί σε πολύ ισχυρή τελική επίδοση στην AMD, που πλησιάζει ή ξεπερνά το 100%, και σε υψηλότερη τελική απόδοση στο SPY (~60–70%) σε σχέση με το γράφημα του τελικού μοντέλου ΤΤΜ. Η TSLA παραμένει ηγετική με μεγάλη σωρευτική απόδοση αλλά και έντονα drawdowns, ενώ η XOM καταλήγει πιο μέτρια. Η AAPL παραμένει κοντά στο μηδέν ή οριακά αρνητική, και η JNJ καταλήγει χαμηλά θετική. Συνολικά, το ARIMA φαίνεται πιο επιθετικό σε ορισμένες μετοχές (ιδίως AMD και SPY), με μεγαλύτερα τόσο τα ανοδικά σκέλη όσο και τις υποχωρήσεις, ενώ το μοντέλο ΤΤΜ δίνει πιο μαζεμένα κέρδη.

### 17) Αποτελέσματα στρατιγικής - διαγράμματα Sharp, Total Return, Max Drawdown

Τα τρία ραβδογράμματα, για κάθε μετοχή χωριστά, συνοψίζουν την απόδοση της ίδιας ακριβώς δοκιμαστικής στρατηγικής που τρέχουμε πάνω στις προβλέψεις του μοντέλου: 1) το Sharpe ratio μετρά πόση απόδοση ανά μονάδα κινδύνου πέτυχε η στρατηγική, 2) το Total return είναι το συνολικό κερδισμένο ποσοστό στο τέλος της περιόδου, ενώ 3) το Max drawdown είναι το μεγαλύτερο ποσοστό πτώσης της καμπύλης πλούτου της στρατηγικής, άρα δείχνει τον χειρότερο ενδιάμεσο «κραδασμό» που θα χρειαζόταν να αντέξουμε.

Το μοτίβο ρίσκου-απόδοσης παραμένει και εδώ, με ελαφρώς διαφορετικές ισορροπίες. Η TSLA και πάλι ηγείται με total return 109% και Sharpe 0.02, αλλά με ακόμη μεγαλύτερο max drawdown (1.42), δείχνοντας ότι το κέρδος απαιτεί μεγάλη αντοχή σε ενδιάμεσες πτώσεις. Η AMD είναι πολύ δυνατή στο ARIMA, με 113% τελικό κέρδος, Sharpe 0.02 και drawdown 1.30, επιβεβαιώνοντας ότι το μονοσήμαντο μοντέλο τάσης ακολουθεί αποτελεσματικά μεγάλές κινήσεις. Ο SPY εμφανίζεται ξανά ως το ισορροπημένο προφίλ με περίπου 72% total return, το υψηλότερο Sharpe στο set (0.04) και χαμηλό σχετικά drawdown (0.39), κάτι που ταιριάζει με την ευρεία διαφοροποίηση του δείκτη. Η PG και η XOM δίνουν μικρομεσαίες αποδόσεις (22% και 10%) με χαμηλά έως μεσαία drawdowns, ενώ η JNJ μένει χαμηλά (6% απόδοση). Η AAPL παραμένει αρνητική (−4%) με drawdown πάνω από 1.0, ένδειξη ότι το απλό ARIMA δεν πιάνει καλά τις φάσεις της στο δείγμα μας.

### 18) Συμπεράσματα

Το ARIMA(0,1,0) αναδεικνύεται σε ισχυρό και λιτό μοντέλο, αλλά εκπαιδεύεται σε περισσότερα δεδομένα χωρίς τεράστια διαφορά από το ΤΤΜ. Επίσης στα παράθυρα γεγονότων το τελικό ΤΤΜ έχει σαφώς καλύτερη επίδοση στην κατεύθυνση. Το ARIMA όμως επέφερε περισσότερα κέρδη στην εξέταση της στρατιγικής. Μετά τη διαφοροποίηση (d=1) η υπόλοιπη γραμμική εξάρτηση είναι περιορισμένη, οπότε οι πρόσθετοι όροι AR/MA σπάνια βελτιώνουν ουσιαστικά τις μετρικές. Προσφέρει στενότερη κατανομή σφαλμάτων (λιγότερες ακραίες τιμές) και αποδίδει πειστικά σε ήρεμες αγορές, αλλά η επίδοσή του υποβαθμίζεται όταν αυξάνει η βραχυπρόθεσμη μεταβλητότητα όπου το ΤΤΜ έχει καλύτερες επιδόσεις .
