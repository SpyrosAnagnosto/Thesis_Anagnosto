### 2) Θερμικοί χάρτες σφαλμάτων
Αρχικά, η μεταφερσιμότητα αξιολογήθηκε με τις τρεις μετρικές απόδοσης **RMSE**, **MAE** και **MAPE** καθεμία εκ των οποίων προσφέρει μια διαφορετική οπτική στην απόδοση του μοντέλου. Οι γραμμές των θερμικών χαρτών αντιπροσωπεύουν σε ποια μετοχή εκπαιδεύτικε το μοντέλο και κάθε στήλη αντιπροσωπεύει την μετοχή στην οποία γίνεται η πρόβλεψη. Η διαγώνιος αντιπροσωπεύει την απόδοση του μοντέλου όταν κάθε μετοχή εκπαιδευεται στον εαυτό της και προβλέπει τον εαυτό της. Όσο πιο σκούρο είναι το χρώμα σε κάθε κουτάκι, τόσο καλύτερη η τιμή της μετρικής. Επίσης, σε κάθε στήλη η καλύτερη τιμή κυκλώνεται με ένα λευκό πλαίσιο.  

**Υπενθύμιση του τι αντιπροσοπεύουν οι μετρικές σφάλματων :**
* Το **RMSE (Ριζική Μέση Τετραγωνική Απόκλιση)** τιμωρεί περισσότερο τα μεγάλα σφάλματα και δίνει έμφαση στη διασπορά των λαθών πρόβλεψης. Αναδεικνύει το μέγεθος και τη μεταβλητότητα των αποτυχιών γενίκευσης.
  
* Το **MAE (Μέση Απόλυτη Απόκλιση)** υπολογίζει τον γραμμικό μέσο όρο των σφαλμάτων, προσφέροντας μια πιο ισορροπημένη εικόνα για τις τυπικές αποκλίσεις των προβλέψεων ανεξαρτήτως κατεύθυνσης.

* Το **MAPE (Μέση Απόλυτη Ποσοστιαία Απόκλιση)** εκφράζει το σφάλμα ως ποσοστό της πραγματικής τιμής, επιτρέποντας συγκρίσεις μεταξύ μετοχών διαφορετικής κλίμακας και δίνει έμφαση στην σχετική ποιότητα πρόβλεψης.

**Διαγώνιος**<br>
Για τις περισσότερες μετοχές (π.χ. XOM, SPY, JNJ, PG), οι διαγώνιες τιμές και στα τρία διαγράμματα είναι από τις χαμηλότερες στη στήλη τους, γεγονός που δείχνει ότι τα μοντέλα τείνουν να αποδίδουν πολύ καλά όταν εκπαιδεύονται στην μετοχή που θα προβλέψουν. Όμως οι τιμές της διαγωνίου δεν είναι πάντα οι καλύτερες, άρα η εκπαίδευση του μοντέλου σε δεδομένα μετοχής που πρόκειται να προβλέψουμε δεν εγγυάται πάντα την καλύτερη απόδοση. 

**Εταιρίες τεχνολογίας**<br>
Τα χειρότερα αποτελέσματα για όλες τις στήλες παράγονται όταν εκπαιδεύουμε σε δεδομένα τεχνολογικών εταιριών (AAPL, TSLA, AMD) ακόμα και όταν προβλέπουν τον εαυτό τους. Αυτό πιθανόν συμβαίνει γιατί στις τιμές αυτών των μετοχών υπάρχει πολύς θόρυβος ο οποίος κατά την εκπαίδευση ενσωματώνεται στα βάρη του ΤΤΜ. Επίσης, τα σφάλματα σε όλες τις μετρικές για την TSLA είναι μεγάλα όταν εκπαιδεύουμε σε οποιαδήποτε μετοχή, επιβεβαιώνοντας την μεγάλη αβεβαιότητα στο πως φέρονται οι μέτοχοι της εταιρίας. 

**Μετοχές σταθερής μεταβλητότητας**<br>
Οι XOM, JNJ και PG πρβλέπουν πολύ καλά η μία την άλλη. Αυτό συμβαίνει γιατί αντιπροσοπεύουν τομείς με υψηλή σταθερότητα (ενέργεια, υγεία, αγαθά πρώτης ανάγκης) που έχουν μικρές μεταβολές τιμών, χαμηλή μεταβλητότητα και ανθεκτικότητα σε οικονομικές κρίσεις. Το ΤΤΜ συλλαμβάνει πολύ καλά τα μοτίβα αυτών των μετοχών. 

Μια ενδιαφέρουσα παρατήρηση είναι όταν χρησιμοποιούνται αυτές οι μετοχές για να προβλεφθει η SPY, έχουν αρκετά χαμηλό MAPE αλλά ψηλά RMSE kai MAE. Όπως είπαμε το MAPE κανονικοποιεί το σφάλμα σε σχέση με την πραγματική τιμή. Έτσι, αποτυπώνει πόσο μεγάλο είναι το σφάλμα σε σχέση με το μέγεθος της ίδιας της μετοχής. Άρα οι XOM, PG, JNJ ενδέχεται να προβλέπουν σωστά την σχετική κίνηση της SPY, οδηγώντας σε χαμηλό MAPE. Τα RMSE και MAE μετρούν το ακατέργαστο μέγεθος του σφάλματος. Η SPY, επειδή είναι ETF που αντιπροσωπεύει τον δείκτη S&P500, έχει υψηλότερη κλίμακα τιμής από τις περισσότερες μεμονωμένες μετοχές. Έτσι, ακόμα κι αν η πρόβλεψη είναι σχετικά κοντά σε ποσοστιαία βάση, αυτές οι μετοχές αδυνατούν να προβλέψουν το πλάτος της SPY σωστά.

Το αντίθετο σθμβαίνει με την XOM. Επειδή είναι μετοχή με τιμές μικρής κλίμακας το ΜΑΡΕ είναι μεγάλο σε όλες τις προβλέψεις. Άρα σε σχέση με την πραγματική τιμή οι προβλέψεις δεν είναι καλές. Παρόλα αυτά τα RMSE και MAE για την προβλεψη της XOM φαίνονται μικρά αλλά αυτό γίνεται μόνο λόγο του ότι η μετοχή έχει γενικά χαμηλές τιμές.


**Η SPY ως ισχυρός γενικευτής**<br>
Σε όλες τις μετρικές σφάλματος, το μοντέλο που εκπαιδεύεται στη SPY εμφανίζει εξαιρετική ικανότητα γενίκευσης, τουλάχιστον ως προς το πλάτος. Το χαμηλό RMSE δείχνει περιορισμένα μεγάλα σφάλματα, χαμηλό MAE, δείχνει συνεπή ακρίβεια και το χαμηλό MAPE, επιβεβαιώνει καλή σχετική απόδοση ανεξαρτήτως κλίμακας. Αυτό υποδηλώνει ότι η SPY ενσωματώνει πιθανόν μια ισορροπημένη αναπαράσταση της αγοράς, επιτρέποντας στο μοντέλο να μεταφέρει αποτελεσματικά τα μαθημένα μοτίβα σε άλλες μετοχές.

#### 2) Θερμικοί χάρτες μετρικών κατεύθυνσης
Συνεχίζουμε την ανάλυση με τις κατευθυντικές μετρικές. Ο πήχης για αυτές τις μετρικές είναι πιο κάτω από άλλους τύπους χρονοσειρών. Η πρόβλεψη της κατεύθυνσης των τιμών στο χρηματιστήριο είναι πολύ δύσκολη και γεμάτη θόρυβο, λόγω της τυχαιότητας και της αστάθειας της αγοράς. Οποιοδήποτε ποσοστό σημαντικά πάνω από το 50% δείχνει κάποια ικανότητα. Μοντέλα που φτάνουν το 60% θεωρούνται πάρα πολύ καλά στον χρηματστηριακό τομέα. Επίσης πάντα πρέπει να προσέχουμε τα εξής σημεία:
- Αν δούμε hit rate πάνω από 60% πρέπει να είμαιστε πολύ καχύποπτοι για overfitting.
- Μοντέλα με υψηλό recall αλλά χαμηλό precision δίνουν πολλά ψευδώς θετικά σήματα.
- Μοντέλα με υψηλό precision αλλά χαμηλό recall ενδέχεται να κάνουν πολλές λανθασμένες προβλέψεις.

**Υπενθύμιση του τι αντιπροσοπεύουν οι μετρικές κατεύθυνσης :**
* Το **Hit Rate** μετρά αν το μοντέλο προέβλεψε σωστά την κατεύθυνση, ανεξαρτήτως του μεγέθους της κίνησης. 

* Το **Precision** μετρά από όλες τις φορές που το μοντέλο προέβλεψε ότι η τιμή θα ανέβει, πόσες φορές είχε δίκιο (false positives).

* Το **Recall** μετρά από όλες τις φορές που η τιμή όντως ανέβηκε, πόσες φορές το προέβλεψε σωστά το μοντέλο (false negatives).

* Το **F1 Score** F1 είναι ο αρμονικός μέσος της ακρίβειας και της ανάκλησης. Εξισορροπεί τις δύο μετρικές, παρέχοντας μια συνολική τιμή απόδοσης.

**Hit Rate**<br>
Οι τιμές Hit Rate κυμαίνονται μεταξύ 0.47 και 0.50. Στην πρόβλεψη της AAPL οι τιμές του Hit Rate είναι σταθερές σε όλα τα μοτέλα. Το ίδιο συμβαίνει και για την PG εκτός της περίπτωσης του μοντέλου που έχει εκπαιδευτεί στην ίδια την PG. Όλα τα μοντέλα παρουσιάζουν καλύτερα Hit Rates στις προβλέψεις των XOM και JNJ. Το μοντέλο που εκπαιδεύτηκε στην XOM έχει τις περισσότερες μέγιστες τιμές. Η γραμμή του μοντέλου που εκπαιδεύτηκε στο SPY δείχνει ομοιόμορφες τιμές Hit Rate (~0.47–0.49), υποδεικνύοντας σταθερή αλλά μέτρια μεταβιβασιμότητα στην κατεύθυνση.

**Precision, Recall, F1**<br>
Σε Precision,Recall και F1, η πρόβλεψη στην SPY εμφανίζει τις υψηλότερες τιμές σχεδόν παντού (συχνά >0.53). Τα μοντέλα που εκπαιδεύτηκαν σε AMD και TSLA εμφανίζουν χαμηλότερα Precision, Recall και F1, ειδικά όταν προβλέπουν άλλες μετοχές. Για τις προβλέψεις της AMD το Precision ειναι μεγαλύτερο από το Recall, πράγμα που σημαίνει πως τα μοντέλα είναι πιο επιθετικά και  προβλέπουν πολλές κινήσεις προς τα πάνω για την AMD (υψηλό recall), αλλά αυτές οι κινήσεις είναι false positives (χαμηλό precision). Τα μοντέλα που εκπαιδεύτηκαν σε ΧΟΜ και η PG έχουν τις περισσότερες μέγιστες τιμές. Τα μοντέλα TSLA και AMD παρουσιάζουν ξανά χαμηλή μεταφερσιμότητα. Το μοντέλο PG ενώ έχει κάποια καλά σκορ, και μερικές φορές τα καλύτερα, δεν έχει συνέπεια (σε μερικές μετοχές έχει τα χαμηλότερα σκορ). Άρα οι XOM και JNJ είναι ισχυροί γενικευτές κατεύθυνσης. Η SPY έχει μικρότερες τιμές από αυτές αλλά έχει σταθερή απόδοση.


### 2) Αξιολόγηση αποτελεσμάτων μοντέλου εκπαιδευμένου στην SPY

Μετά τα αποτελέσματα καλύτερου ποσοστού εκπαίδευσης και της παραπάνω ανάλυσης εκπαιδεύουμε το μοντέλο στo 42% της SPY, δηλαδή σε 1600 γραμμές δεδομένων 

Η εκπαίδευση στη SPY φαίνεται να μεταφέρει καλά μοτίβα γενικής αγοράς σε σταθερές και μέτρια μεταβλητές μετοχές (JNJ, PG, XOM, AAPL). Σε τίτλους με υψηλή beta (TSLA, AMD), η βελτίωση έναντι του zero-shot υπάρχει αλλά είναι περιορισμένη, καθώς η SPY δεν περιέχει αρκετές ακραίες κινήσεις για να μάθει το μοντέλο να τις χειρίζεται. Η βελτίωση σε σχέση με το zero-shot είναι εμφανής κυρίως στην προβλεψιμότητα της τάσης και στη μείωση συστηματικών αποκλίσεων σε σταθερά περιβάλλοντα.

#### Ανάλυση στο test set
* **AAPL**: Το μοντέλο ακολουθεί πολύ καλά την τάση και τις βραχυπρόθεσμες κινήσεις. Υπάρχει μικρή υποεκτίμηση στις κορυφές και στις έντονες ανόδους, αλλά συνολικά η προβλεπόμενη πορεία είναι κοντά στην πραγματική.
* **TSLA**: Παρότι η γενική κατεύθυνση αποδίδεται σωστά, σε περιόδους εκρηκτικής ανόδου ή πτώσης το μοντέλο υποεκτιμά συστηματικά τα άκρα. Η γνώση από τη SPY δεν αρκεί για να καλύψει την ακραία μεταβλητότητα της TSLA.
* **XOM**: Καλή απόδοση και διατήρηση της σωστής κατεύθυνσης, με σχετικά μικρές αποκλίσεις, ακόμα και σε περιόδους αυξημένης μεταβλητότητας λόγω ενεργειακών γεγονότων.
* **SPY**: (ίδια μετοχή) Σχεδόν τέλεια προσαρμογή στο test set, με ελάχιστες αποκλίσεις, όπως θα αναμέναμε λόγω εκπαίδευσης στην ίδια κατανομή.
* **JNJ**: Εξαιρετική εφαρμογή, με μεγιστες αποκλίσεις της τάξης των 10 δολλαρίων. Η σταθερότητα της μετοχής ευνοεί τη μεταφορά γνώσης από τη SPY και οδηγεί σε σχεδόν επικαλυπτόμενες καμπύλες πρόβλεψης-πραγματικότητας.
* **AMD**: Σε περιόδους έντονης ανόδου και πτώσης διατηρεί την κατεύθυνση, αλλά υποτιμά το μέγεθος των κινήσεων. Η εκπαίδευση στη SPY βοηθά στη σταθεροποίηση της πρόβλεψης, αλλά η υψηλή μεταβλητότητα της AMD παραμένει πρόκληση.
* **PG**: Παρόμοια με την JNJ, η πρόβλεψη είναι σταθερή και με μικρές αποκλίσεις. Η χαμηλή μεταβλητότητα της PG ταιριάζει με τη συμπεριφορά που το μοντέλο έμαθε από τη SPY.

#### Ανάλυση ανά γεγονός
Κατα μέσο όρο έχουμε καλύτερη απόδοση αλλά σε όλα τα παράθυρα παρατηρείται το ίδιο πρόβλημα. Όταν υπάρχει μια πτώση τιμής το μοντέλο θεωρεί πως αυτή η πτώση είναι πολύ μεγαλύτερη από την πραγματικότητα.

- AAPL – Κατάρρευση & Ανάκαμψη (2020-03-10)
Το μοντέλο αποτυπώνει ικανοποιητικά τη γενική κατεύθυνση τόσο της πτωτικής όσο και της ανοδικής φάσης, ωστόσο υποεκτιμά το μέγεθος της πτώσης και της επακόλουθης ανάκαμψης. Η εκπαίδευση στη SPY φαίνεται να μειώνει την καθυστέρηση στην αντίδραση, αλλά δεν αποτρέπει πλήρως τις αποκλίσεις στο εύρος των κινήσεων.

- TSLA – Αποφόρτιση Μετοχής Υψηλού Beta (2021-01-15)
Η πρόβλεψη αναγνωρίζει την ανοδική πορεία πριν την κορύφωση, αλλά δυσκολεύεται να προσαρμοστεί στην πτωτική φάση. Οι κορυφές και οι έντονες διορθώσεις υποεκτιμώνται, ενώ η μετάβαση από άνοδο σε πτώση δεν εντοπίζεται έγκαιρα, κάτι αναμενόμενο λόγω της ηπιότερης μεταβλητότητας της SPY.

- XOM – Κορύφωση Πετρελαϊκού Κύκλου (2022-06-01)
Η γενική πορεία και η χρονική ευθυγράμμιση είναι καλές, αν και οι τιμές προβλέπονται σταθερά ελαφρώς χαμηλότερες από τις πραγματικές. Η σταθερότητα της πρόβλεψης δείχνει ότι η εκπαίδευση στη SPY μεταφέρει χρήσιμα κυκλικά μοτίβα και σε άλλες ενεργειακές μετοχές.

- SPY – Πτωτική Αστάθεια με Εναλλαγές Κατεύθυνσης (2022-09-15)
Η καθοδική τάση αναγνωρίζεται σωστά, αλλά η δομή και το μέγεθος των ενδιάμεσων κινήσεων αποτυπώνονται με αστοχίες. Τα αποτελέσματα είναι χειρότερα από το απλό μοντέλο. Ίσως έχουμε κάποια βαθμό υπερπροσαρμογής για την πρόβλεψη της SPY.

- JNJ – Περίοδος Χαμηλής Μεταβλητότητας (2019-08-01)
H πρόβλεψη του γεγονότος έχει προβλήματα καθώς το μοντέλο δεν αντιλαμβάνεται τη σταθερότητα της περιόδου. Θεωρεί πως μικρές πτώσεις είναι πολύ μεγαλύτερες και μετά προσπαθεί να ξαναφτάσει την πραγματική τιμή.

- AMD – Μαζική Πώληση Τεχνολογικών Μετοχών (2018-10-10)
Η πρόβλεψη εμφανίζει καθυστέρηση στην προσαρμογή και υποεκτίμηση του εύρους των κινήσεων, ιδιαίτερα στις έντονες φάσεις πτώσης και ανάκαμψης. Ωστόσο, η εκπαίδευση στη SPY μειώνει κάπως τη διασπορά των σφαλμάτων σε σταθερότερες περιόδους.

- PG – Ηρεμία Ανεπηρέαστη από Μακροοικονομικούς Παράγοντες (2015-06-15)
Η πρόβλεψη είναι χειρότερη απτό απλό μοντέλο. Και εδώ παρατηρείται η τάση να θεωρεί πως οι πτώσεις είναι πολύ μεγάλες.

#### Μετρικές στο test set
Οι μετοχές XOM, PG και JNJ παρουσιάζουν πολύ χαμηλό σφάλμα, με MAE περίπου 2.4–2.6 και RMSE κοντά στο 3.1–3.3. Επίσης, το χαμηλό MAPE σε αυτές δηλώνει πως η διαφορετική κλίμακα δεν επηρεάζει αρνητικά. Το F1 score πάνω από 0.51 υποδηλώνει εξαιρετική ισορροπία ανάμεσα στην ακρίβεια (precision) και την ανάκληση (recall). Το hit rate για τη JNJ ξεπερνά το 0.50.

Όσον αφορά τον δείκτη SPY, όπου έγινε εκπαίδευση και πρόβλεψη στον ίδιο τίτλο, τα αποτελέσματα είναι αναμενόμενα πολύ καλά. Το MAE είναι 2.94, το RMSE 3.86 και το MAPE 2.15. Το F1 score φτάνει το 0.528, δείχνοντας ισχυρή απόδοση σε precision και recall.
