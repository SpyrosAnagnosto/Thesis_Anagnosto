# Πείραμα 5 - Βελτιστοποίηση υπερπαραμέτρων/Ανάλυση Σημαντικότητας Χαρακτηριστικών
#### Διπλωματική εργασία - Αναγνωστόπουλος Σπύρος

### 1) Περιγραφή πειράματος
Το μικτο μοντέλο έδειξε πολύ καλή απόδοση στο πλάτος των κινήσεων. Παρ΄ολα αυτά η προβλέψεις εξακολουθούν να υστερούν χρονικά. Σε αυτό το πείραμα θα προσπαθήσουμε να βελτιστοποιήσουμε το μοντέλο. Μέχρι τώρα, διαλέγαμε τις υπερπαραμέτρους *learning rate* και *batch size* στην τύχη. Επίσης, μέχρι τώρα εκπαιδεύαμε το μοντέλο με είσοδο μόνο στην τιμή κλεισίματος. Πρώτα θα βρούμε τις καλύτερες τιμές για τις υπερπαραμέτρους. Ύστερα θα μελετήσουμε την σημαντικότητα των χαρακτηριστικών (feature importance). Έκτός από τις μεταβλητές που βρίσκονται ήδη στα δεδομένα προσθέτουμε και όλους τους τεχνικούς δείκτες (βλ. θεωρητικό υπόβαθρο). 



#### Ερευνητικά Ερωτήματα Πειράματος 5 – Βελτιστοποίηση υπερπαραμέτρων/Ανάλυση Σημαντικότητας Χαρακτηριστικών:
- Τι ρόλο παίζει το learning rate και το batch size στην επίδοση του μοντέλου;

- Υπάρχει ιδανικό learning rate και batch size που θα βελτιώσουν σημαντικά την απόδοση;

- Πόσο βοηθάνε οι μεταβλητές open, high, low, volume στην πρόβλεψη;

- Υπάρχουν τεχνικοί δείκτες (βλ. θεωριτικό υπόβαθρο) που αν χρησιμοποιηθούν ως είσοδος στην εκπαίδευση θα κάνουν την πρόβλεψη ακόμα καλύτερη;






### 2) Βελτιστοποίηση υπερπαραμέτρων
Σε αντίθεση με τις παραμέτρους του μοντέλου (βάρη), που μαθαίνονται κατά την εκπαίδευση, οι υπερπαράμετροι καθορίζονται πριν από την εκπαίδευση. Οι υπερπαράμετροι ενός μοντέλου έχουν καθοριστικό ρόλο στην επίδοση του. Μερικές επιλογές υπερπαραμέτρων εξηγήθηκαν στα προηγούμενες ενότητες. Εδώ επικεντρωνόμαστε στο learning rate και στο batch size. 

Για την βελτιστοποίηση θα χρησημοποιήσουμε το εργαλείο Optuna. Το Οptuna είναι είναι ένα εργαλείο βελτιστοποίησης υπερπαραμέτρων ανοιχτού κώδικα το οποίο διακρίνεται για την ταχύτητα, καθώς χρησιμοποιεί έξυπνη δειγματοληψία και την αποτελεσματικότητα, υποστηρίζοντας την περικοπή μη υποσχόμενων δοκιμών για εξοικονόμηση χρόνου. Αντί να δοκιμάζει όλους τους συνδυασμούς (όπως το Grid Search), το Optuna ορίζει έναν χώρο αναζήτησης, προτείνει πολλά υποσχόμενες υπερπαραμέτρους, τις αξιολογεί και μαθαίνει από προηγούμενες δοκιμές για να βελτιώσει τις μελλοντικές προτάσεις. Είναι ιδιαίτερα χρήσιμο για σύνθετα μοντέλα ή μεγάλα σύνολα δεδομένων, όπου η εξαντλητική βελτιστοποίηση δεν είναι εφικτή.

Για να χρησιμοποιήσουμε το Optuna για τη ρύθμιση ενός μοντέλου, πρέπει πρώτα να ορίσουμε μια **συνάρτηση στόχο**. Αυτή η συνάρτηση:
1) λαμβάνει τις προτάσεις δοκιμών από το Optuna
2) διαμορφώνει το μοντέλο χρησιμοποιώντας τις προτεινόμενες υπερπαραμέτρους
3) το εκπαιδεύει και το αξιολογεί,
4) επιστρέφει ένα αριθμητικό σκορ προς ελαχιστοποίηση (σε αυτήν την περίπτωση, την eval loss).

Στη συνέχεια, δημιουργείται μια **μελέτη Optuna**:
1) πραγματοποιούμε τη μελέτη για 10 ξεχωριστές δοκιμές. Σε κάθε δοκιμή, το Optuna επιλέγει έναν διαφορετικό συνδυασμό υπερπαραμέτρων. 
2) το μοντέλο εκπαιδεύεται και αξιολογείται, και το αποτέλεσμα καταγράφεται σε ένα αρχείο.
3) μετά από όλες τις δοκιμές, το Optuna εντοπίζει το καλύτερο σύνολο υπερπαραμέτρων που οδήγησε στη χαμηλότερη απώλεια επικύρωσης.

#### Διάγραμμα Ιστορικού Βελτιστοποίησης

Το διάγραμμα οπτικοποιεί τον τρόπο με τον οποίο αλλάζει το eval loss κατά τη διάρκεια κάθε δοκιμής στη μελέτη του Optuna. Κάθε μπλε τελεία αντιπροσωπεύει μια δοκιμή με έναν συγκεκριμένο συνδυασμό υπερπαραμέτρων, και η κόκκινη γραμμή παρακολουθεί την καλύτερη απώλεια επικύρωσης που επιτεύχθηκε μέχρι στιγμής.


Η χαμηλότερη απώλεια επικύρωσης που επιτεύχθηκε ήταν περίπου 0.447, ορατή στην τελική δοκιμή. Η βελτιστοποίηση βελτιωνόταν με την πάροδο του χρόνου, με την κόκκινη γραμμή να κατεβαίνει, ιδιαίτερα μεταξύ των δοκιμών 2 και 3. Οι πρώτες δοκιμές είχαν μεγαλύτερη μεταβλητότητα και υψηλότερη απώλεια, δείχνοντας ότι ο αλγόριθμος εξερεύνησε χειρότερες διαμορφώσεις πριν συγκλίνει σε καλύτερες. Αυτό υποδεικνύει ότι το Optuna εξερεύνησε με επιτυχία τον χώρο παραμέτρων και βελτίωσε την απόδοση του μοντέλου με την πάροδο του χρόνου επιλέγοντας καλύτερες υπερπαραμέτρους.

Στην τελική δοκιμή οι υπερπαραμετροι ήταν: **learning_rate = 0.0022490307997647717** και **batch_size = 64**.

#### Σημαντικότητες Υπερπαραμέτρων
Αυτό το ραβδόγραμμα δείχνει τη σχετική σημασία κάθε υπερπαραμέτρου στην ελαχιστοποίηση της συνάρτησης στόχου. Αυτή υπολογίζεται χρησιμοποιώντας μια προσέγγιση σημαντικότητας χαρακτηριστικών βασισμένη σε κάποιο μοντέλο (π.χ. δάση τυχαίων αποφάσεων).

* Το learning_rate συμβάλλει στο ~89% της διαφοράς στην απόδοση.

* Το batch_size συμβάλλει πολύ λιγότερο (~11%).

Αυτό μας λέει ότι η ρύθμιση του ρυθμού μάθησης είχε σημαντικά μεγαλύτερο αντίκτυπο στην απόδοση του μοντέλου από ό,τι η ρύθμιση του μεγέθους batch. Το μοντέλο είναι εξαιρετικά ευαίσθητο στον ρυθμό μάθησης, ίσως οι κακές τιμές σε μετρικές κατεύθυνσης να ευθύνονταν στην επιλογή αυτής της υπερπαραμέτρου.

#### Διάγραμμα Παράλληλων Συντεταγμένων

Το Διάγραμμα Παράλληλων Συντεταγμένων εμφανίζει τις σχέσεις μεταξύ των υπερπαραμέτρων και των προκυπτουσών τιμών στόχου σε όλες τις δοκιμές. Κάθε γραμμή αντιπροσωπεύει μια δοκιμή.
* Αριστερός Άξονας: Τιμή στόχος (eval loss).
* Μεσαίος Άξονας: batch_size (κατηγορικό: 32, 64, 128).
* Δεξιός Άξονας: learning_rate (λογαριθμική κλίμακα).
  
Οι γραμμές χρωματίζονται ανάλογα με την απόδοση, με τις πιο σκούρες γραμμές να αντιπροσωπεύουν χαμηλότερη απώλεια. Οι χαμηλότερες απώλειες επιτυγχάνονται όταν το learning_rate είναι περίπου 0.0022. Πολύ χαμηλοί ρυθμοί μάθησης (π.χ. 0.0001) ή πολύ υψηλοί (π.χ. 0.0025+) υποβαθμίζουν την απόδοση.

Δεν υπάρχει σαφής τάση στο batch_size. Kαλή απόδοση παρατηρείται σε όλες τις τιμές, επιβεβαιώνοντας ό,τι έδειξε το διάγραμμα σημαντικότητας.

### 3) Ανάλυση σημαντικότητας χαρακτηριστικών
Η ανάλυση σημαντικότητας χαρακτηριστικών μας βοηθάει να καταλάβουμε ποιές μεταβλητές εισόδου έχουν μεγαλύτερη προβλεπτική αξία. Χρησιμοποιόυνται δύο τρόποι:

* **SHAP (shapley additive explanations):** Ποσοτικοποιεί την συνεισφορά κάθε μεταβλητής σε μεμονωμένες προβλέψεις. Είναι βασισμένο στη θεωρία παιγνίων και είναι ανεξάρτητο του μοντέλου που χρησιμοποιείται.
  
* **Σημαντικότητα μεταθέσεων :** Ανακατεύει με τυχαίο τρόπο μεταβλητές εισόδου και μετράει την επίδοση του μοντέλου. Μεγάλη πτώση στην απόδοση δείχνει πως η μεταβλητή που ανακατεύτηκε ήταν σημαντική.   

#### Διάγραμμα SHAP 
Το ραβδόγραμμα απεικονίζει τις μέσες τιμές σημαντικότητας χαρακτηριστικών SHAP που συγκεντρώθηκαν από προβλέψεις σε όλες τις μετοχές. Οι τιμές SHAP παρέχουν πληροφορίες σχετικά με το πόσο συμβάλλει κάθε χαρακτηριστικό στις προβλέψεις του μοντέλου, με υψηλότερες τιμές να υποδηλώνουν μεγαλύτερη επιρροή.

Τα χαρακτηριστικά των ακατέργαστων τιμών κυριαρχούν σε σημασία:
* Η χαμηλότερη ημερήσια τιμή **low** είναι το πιο επιδραστικό χαρακτηριστικό συνολικά, με μέση τιμή SHAP 30.25.
* Η υψηλότερη ημερήσια τιμή **high** και η τιμή ανοίγματος **open** ακολουθούν στενά, υποδηλώνοντας ότι τα βασικά επίπεδα τιμών φέρουν το ισχυρότερο προγνωστικό σήμα στο μοντέλο.

Οι τεχνικοί δείκτες παρουσιάζουν ποικίλους βαθμούς επιρροής:
* Μεταξύ των κινητών μέσων όρων, ο σταθμισμένος κινητός μέσος όρος 20 ημερών **wma_20** και ο εκθετικός κινητός μέσος όρος **ema_20** συμβάλλουν σημαντικά στις προβλέψεις του μοντέλου.
* Ο τριπλός εκθετικός κινητός μέσος όρος **tema_20** έχει επίσης αξιοσημείωτο, αν και χαμηλότερο, μέσο αντίκτυπο.
* Αντίθετα, ο απλός κινητός μέσος όρος **sma_20** και η άνω ζώνη Bollinger **bb_upper** παρουσιάζουν μικρή αλλά όχι μηδενική επιρροή.

#### Διαγραμμάτα Beeswarm

Τα διαγράμματα beeswarm προσφέρουν έναν διαισθητικό τρόπο για να καταλάβουμε πώς κάθε χαρακτηριστικό στο σύνολο δεδομένων συμβάλλει στις προβλέψεις του μοντέλου. Κάθε χαρακτηριστικό αναγράφεται στον κάθετο άξονα. Κάθε σημείο στο διάγραμμα αντιπροσωπεύει μια γραμμή από το σύνολο δεδομένων της μετοχής.

Ο οριζόντιος άξονας αντιπροσωπεύει την τιμή SHAP, η οποία μας λέει πόσο συμβάλλει ένα χαρακτηριστικό στην έξοδο του μοντέλου για ένα συγκεκριμένο σημείο δεδομένων. Τα σημεία στα δεξιά (θετικές τιμές SHAP) σημαίνουν ότι το χαρακτηριστικό αύξησε την προβλεπόμενη τιμή. Τα σημεία στα αριστερά (αρνητικές τιμές SHAP) υποδεικνύουν ότι το χαρακτηριστικό μείωσε την πρόβλεψη. Για παράδειγμα, αν προβλέπουμε μια τιμή μετοχής και ένα σημείο δεδομένων βρίσκεται πολύ στα δεξιά, σημαίνει ότι αυτό το χαρακτηριστικό ώθησε το μοντέλο προς την πρόβλεψη μιας υψηλότερης τιμής για αυτό το δείγμα.

Το πάχος της γραμμής σε οποιοδήποτε σημείο του οριζόντιου άξονα δείχνει πόσο συχνή είναι αυτή η τιμή SHAP σε όλο το σύνολο δεδομένων. Οι πιο παχιές περιοχές υποδηλώνουν ότι πολλά σημεία δεδομένων είχαν παρόμοιες τιμές SHAP δηλαδή το επίπεδο αντίκτυπου του χαρακτηρηστικού σε αυτά είναι κοινό. Οι πιο λεπτές περιοχές σημαίνουν ότι λιγότερα σημεία δεδομένων είχαν αυτή την ποσότητα επιρροής από το χαρακτηριστικό.

Το χρώμα κάθε σημείου λέει την αρχική τιμή του χαρακτηριστικού για αυτό το σημείο δεδομένων. Το κόκκινο αντιπροσωπεύει υψηλές αρχικές τιμές χαρακτηριστικού. Οι μπλε τελείες αντιπροσωπεύουν χαμηλές τιμές. Αυτό μας επιτρέπει να δούμε όχι μόνο πόσο επηρέασε ένα χαρακτηριστικό μια πρόβλεψη, αλλά και προς ποια κατεύθυνση η τιμή του επηρέασε το μοντέλο.

#### Σημαντικότητα μεταθέσεων :
Η σημαντικότητα μεταθέσεων (permutation importance) είναι μια τεχνική που χρησιμοποιείται για να μετρηθεί πόσο σημαντικό είναι κάθε χαρακτηριστικό για ένα εκπαιδευμένο μοντέλο. Ένα χαρακτηριστικό ανακατεύεται τυχαία σε ολόκληρο το σύνολο δεδομένων. Η απόδοση του μοντέλου επαναξιολογείται. Αν η απόδοση μειωθεί σημαντικά, αυτό σημαίνει ότι το συγκεκριμένο χαρακτηριστικό ήταν σημαντικό. Αν η απόδοση δεν αλλάξει ιδιαίτερα, το χαρακτηριστικό πιθανότατα δεν ήταν σημαντικό.

Υπολογίζω την σημαντικότητα μεταθέσεων για κάθε μετοχή και μετά παίρνω των μέσο όρο όλων για κάθε χαρακτηρηστικό. Στον κάθετο άξονα είναι τα ονόματα των χαρακτηριστικών και στον οριζόντιο άξονα η μείωση στην απόδοση του μοντέλου.

* **wma\_20 = 22.682,  tema\_20 = 5.500, ema\_20 = 3.685** :<br>
Ο σταθμισμένος κινητός μέσος ορος (20 ημερών) κυριαρχεί. Η ανακατάταξή του βλάπτει σοβαρά την πρόβλεψη. Ο τριπλός εκθετικός κινητός μέσος όρος, δείχνει ότι το μοντέλο βασίζεται έντονα στη κατεύθυνση της τάσης. Ο κλασικός εκθετικός κινητός μέσος επιβεβαιώνει την εξάρτηση του μοντέλου από στρατηγικές παρακολούθησης τάσεων.

* **high = 1.519, bb_upper = 962 και low, open, sma_20 ≈ 600–800**<br>
Η ημερήσια υψηλή τιμή βοηθά στις προβλέψεις, αλλά με όχι τόσο ισχυρό τρόπο όσο οι δείκτες τάσης. Τα υπόλοιπα τέσσερα χαρακτηριστικά είναι χρήσιμα αλλά δευτερεύοντα.

* **rsi\_14, atr\_14, obv, cci\_14, cmf\_20, volume, stoch\_k ≈ 0 ή αρνητικά**
Αυτά τα χαρακτηριστικά είτε δεν βοηθούν ιδιαίτερα είτε προσθέτουν θόρυβο στο μοντέλο. Η ανακατάταξή τους δεν μειώνει την απόδοση σε κάποιες περιπτώσεις μπορεί να τη βελτιώνει ελαφρώς.

#### Σχόλιο για τον όγκο
Στα αποτελέσματα των παραπάνω αναλύσεων ο όγκος (ο αριθμός συναλλαγών), που θα περιμέναμε να παίζει σημαντικό ρόλο, συνεισφέρει ελάχιστα στις προβλέψεις του μοντέλου σε σύγκριση με άλλα χαρακτηριστικά. Αυτό μπορεί να συμβαίνει επειδή ο όγκος μπορεί έχει μεγάλες μεταβολές ανά μέρα λόγω διάφορων γεγονότων (ειδήσεις, νομοθετικές πράξεις κ.τ.λ). ### 4) Συμπέρασμα
Το **learning rate** είναι πολύ σημαντικό για την απόδοση του ΤΤΜ. Το **wma_20** είναι το πιο επιδραστικό χαρακτηριστικό, συχνά σηματοδοτώντας αντιστροφές. Οι τιμές OHLC (**low, high, open**) κυριαρχούν στις ευμετάβλητες μετοχές (AAPL, TSLA, AMD). Οι δείκτες έχουν μεγαλύτερη σημασία σε σταθερές μετοχές. Τα χαρακτηριστικά που βασίζονται στον όγκο συμβάλλουν σχεδόν καθόλου.

**Επιλογές για το τελικό μοντέλο: learning rate = 0.00225,  open, high, low, wma_20, ema_20, tema_20, sma_20**
