### 1) Περιγραγή πειράματος
Αρχικά, ελέγξαμε το ΤΤΜ στην πιο απλή περίπτωση χωρίς καμμία εκπαίδευση. Στο πείραμα 3 βρήκαμε το καλύτερο ποσοστό εκπαίδευσης στο 42%. Μετα περάσαμε στην ανάλυση μεταφερσιμότητας μεταξύ των μεταξύ των μετοχών και αποφασίσαμε να χρησιμοποιήσουμε για εκπαίδευση αρχικά τον δεικτη SPY και μετά την ανάλυση με συσταδοποίηση και γράφους τις SPY, JNJ, AAPL, XOM, δηλαδή ένα σύνολο εκπαίδευσης που δεν περιλαμβάνει την πιο δύσκολη σε πρόβλεψη μετοχή TSLA. Τέλος στο προηγούμενο πείραμα κάναμε βελτιστοποίηση παραμέτρων και ανάλυση σημαντικότητας χαρακτηριστικών. Έτσι σε αυτήν την ενότητα εκπαιδεύουμε και αξιολογούμε το τελικό (καλύτερο) μοντέλο ΤΤΜ.

Τελικό μοντέλο:
- Εκπαιδευμένο στο 42% των SPY, JNJ, AAPL, XOM
- learning rate =  0.00225
- batch size = 64
- Features = \[open,
              high,
              low,
              wma_20,
              ema_20,
              tema_20,
              sma_20
             ]
- Epochs 5
- προβλεψη στην Close

### 2) Οπτική αξιολόγηση του τελικού μοντέλου

Σαφώς η καλύτερη επίδοση που έχουμε δει μέχρι τώρα. Το μοντέλο γενικεύει πολύ καλά σε περιόδους χαμηλής και μέσης μεταβλητότητας (SPY, JNJ, PG, XOM), όπου οι γραμμές πρακτικά συμπίπτουν. Σε μετοχές υψηλού beta (TSLA, AMD) η κατεύθυνση συλλαμβάνεται, αλλά το πλάτος κινήσεων τείνει λίγο να υποεκτιμάται και τα σημεία καμπής συνοδεύονται από μικρή, αλλά ορατή, υστέρηση. Αυτό υποδηλώνει έναν βαθμό εξομάλυνσης της πρόβλεψης, ο οποίος μπορεί να είναι καλός για τη σταθερότητα αλλά μπορεί να κάνει λάθη σε ακραία επεισόδια.

### 3) Κατανομή σφάλματος πρόβλεψης σε σύγκριση με τα προηγούμενα προεκπαιδευμένα μοντέλα
Παρακάτω παρουσιάζουμε τα διαγράμματα των y_true - y_pred για τα 3 προσαρμοσμένα μοντέλα. Αντί να κάνουμε, όπως στο zeroshot, ξεχωριστό διάγραμμα για κάθε μετοχή παίρνουμε τον μέσο όρο για όλες τις μετοχές για να αποκτήσουμε μια γενική εικόνα. 

**Μοντέλο SPY** <br>
Η κατανομή είναι πλατιά με έντονες δεξιές ουρές (πολύ μεγάλα θετικά σφάλματα). Αυτό υποδηλώνει συχνές υποεκτιμήσεις σε απότομα ανοδικά κινήματα και γενικά μεγαλύτερη διασπορά σφαλμάτων. Το κέντρο φαίνεται ελαφρώς μετατοπισμένο προς θετικές τιμές, ένδειξη μικρής συστηματικής υποεκτίμησης.

**Μικτό σετ (SPY–JNJ–AAPL–XOM)** <br>
Η διασπορά συρρικνώνεται αισθητά και οι ουρές περιορίζονται. Η κορυφή γύρω από το μηδέν είναι σαφώς υψηλότερη, πράγμα που σημαίνει ότι οι περισσότερες προβλέψεις πέφτουν πολύ κοντά στην πραγματική τιμή. Παραμένει μια ήπια ασυμμετρία στις ουρές, αλλά οι ακραίες τιμές μειώνονται σημαντικά σε σχέση με το σκέτο SPY.

**Τελικό μοντέλο** <br>
Η πυκνότητα είναι πλέον στενή, οξυκόρυφη και σχεδόν συμμετρική γύρω από το μηδέν. Οι ουρές είναι αισθητά κομμένες, δείχνοντας σαφή μείωση  των μεγάλων σφαλμάτων. Το μοντέλο σπάνια κάνει μεγάλες υπερεκτιμήσεις/υποεκτιμήσεις.

Ενδεικτικά, η PG έχει MAE=1.24 και RMSE=1.71 με MAPE<1.0%, ενώ οι JNJ/XOM κινούνται στον ίδιο, πολύ χαμηλό, θόρυβο (MAE=1.3, RMSE=1.6–1.8). Η AAPL βρίσκεται ένα σκαλί πιο πάνω (MAE=2.13, RMSE=2.86). * Στο καθεστώς **Χαμηλής Μεταβλητότητας** είναι η καλύτερη συμπεριφορά (MAE≈1.29, RMSE≈1.76, MAPE≈0.97%) με F1≈0.523. Το μοντέλο αποδίδει σταθερά όταν οι διακυμάνσεις είναι μικρές και οι τάσεις ομαλές.

### 6) Ομαδικά διαγράμματα σύγκρισης μετρικών
Στο αριστερό πάνελ (σφάλματα) το τελικό μοντέλο παρουσιάζει τις χαμηλότερες μέσες τιμές και στις τρεις μετρικές (MAE, RMSE, MAPE). Η μείωση έναντι του απλού SPY είναι έντονη (τάξης \~50% για MAE/RMSE και ακόμη μεγαλύτερη για MAPE), ενώ υπερτερεί σταθερά και του πολυ μετοχικού μοντέλου χωρίς μεταβλητές εισόδου με επιπλέον \~10–20% βελτίωση. Αυτό δείχνει ότι το τελικό μοντέλο αποτυπώνει καλύτερα τόσο το επίπεδο όσο και τη διακύμανση της τιμής στο test set. Στο δεξί πάνελ (κατευθυντικές μετρικές) οι διαφορές είναι αναμενόμενα μικρότερες, καθώς οι δείκτες αυτοί κινούνται στο στενό εύρος 0.47–0.53. Το τελικό μοντέλο διατηρεί ήπιο πλεονέκτημα σε Recall και F1, με Precision και Hit Rate ουσιαστικά ισοδύναμα με τα άλλα δύο μοντέλα. Με άλλα λόγια, το τελικό μοντέλο βελτιώνει αισθητά την ακρίβεια τιμολόγησης χωρίς να θυσιάζει τη σταθερότητα της κατεύθυνσης.

### 7) Διαγράμματα κυλιόμενης μεταβλητότητας - σφάλματος

Συστηματικά, το πράσινο νέφος είναι πιο συμπιεσμένο κοντά στον οριζόντιο άξονα και με μικρότερη διασπορά καθώς αυξάνεται η μεταβλητότητα. Αυτό δείχνει ότι το τελικό μοντέλο διατηρεί χαμηλότερα σφάλματα όχι μόνο σε ήπιες συνθήκες αλλά και όταν η βραχυπρόθεσμη μεταβλητότητα ανεβαίνει. Το μπλε νέφος (SPY-only) παρουσιάζει τη μεγαλύτερη κάθετη διασπορά και περισσότερες ακραίες τιμές, ιδίως στη δεξιά ουρά (υψηλή μεταβλητότητα), ένδειξη ισχυρής ευαισθησίας των σφαλμάτων. Το πορτοκαλί (μικτό μίας μεταβλητής εισόδου) βρίσκεται ενδιάμεσα, βελτιώνοντας αισθητά έναντι του SPY-only αλλά χωρίς να φτάνει τη σταθερότητα του τελικού.

Οι μετοχές υψηλού beta (TSLA, AMD) αναδεικνύουν καθαρά αυτή τη διαφοροποίηση. Στα ίδια επίπεδα μεταβλητότητας, το καλύτερο μοντέλο εμφανίζει αισθητά χαμηλότερα σφάλματα και σαφώς λιγότερες ακραίες τιμές, ενώ το SPY-only απλώνεται σε μεγάλα σφάλματα. Στις πιο αμυντικές (JNJ, PG) η διαφορά μεταξύ μοντέλων παραμένει υπέρ του τελικού, αν και είναι φυσιολογικά μικρότερη λόγω του στενότερου εύρους μεταβλητότητας. Για την XOM στο τελικό, το νέφος παραμένει χαμηλό σε όλο το εύρος, στοιχείο συνέπειας σε περιόδους κυκλικής μεταβολής.

### 9) Αποτελέσματα σε παράθυρα γεγονότων

**AAPL – «Κατάρρευση & Ανάκαμψη» (2020-03-10).**
Η πορεία κατάρρευσης και η μετέπειτα άνοδοσ αποτυπώνονται σωστά. Η πρόβλεψη ακολουθεί στενά τη φάση της πτώσης και επανέρχεται ομαλά στην ανοδική φάση, με μικρή χρονική υστέρηση γύρω από το τοπικό ελάχιστο και ήπια υποτίμηση του πλάτους στην κορύφωση της ανάκαμψης.

**TSLA – «Αποφόρτιση Μετοχής Υψηλού Beta» (2021-01-15).**
Το μοντέλο καταγράφει ικανοποιητικά την προ της κορύφωσης άνοδο και την επακόλουθη πτώση, παραμένοντας ελαφρώς απαισιόδοξο πριν την άνοδο και ελφρως αισιόδοξο μετά την πτώση. Υπάρχει εξομάλυνση του πλάτους και μικρές αποκλίσεις στην αρχή της πτωτικής φάσης.

**XOM – «Κορύφωση Πετρελαϊκού Κύκλου» (2022-06-01).**
Η πρόβλεψη είναι χρονικά ευθυγραμμισμένη και ακολουθεί τη μεταστροφή του κύκλου με περιορισμένη απόκλιση στο επίπεδο. Η άνοδος πριν την κορύφωση και η ομαλή διόρθωση στη συνέχεια αναπαράγονται με καλή ακρίβεια, στοιχείο που υποδηλώνει ότι τα κυκλικά μοτίβα έχουν εσωτερικευτεί.

**SPY – «Πτωτική Αστάθεια με Εναλλαγές Κατεύθυνσης» (2022-09-15).**
Η καθοδική τάση με διαδοχικές αναπηδήσεις αποδίδεται συνολικά σωστά. Το μοντέλο τείνει να στρογγυλεύει τις αιχμές, με μικρή υστέρηση στα τοπικά υψηλά/χαμηλά, αλλά η κατεύθυνση και ο ρυθμός πτώσης είναι πειστικοί.

**JNJ – «Περίοδος Χαμηλής Μεταβλητότητας» (2019-08-01).**
Η επικάλυψη πραγματικής-προβλεπόμενης τιμής είναι σχεδόν πλήρης. Το μοντέλο παραμένει καλά βαθμονομημένο σε όλο το διάστημα, με αποκλίσεις οριακού μεγέθους.

**AMD – «Μαζική Πώληση Τεχνολογικών Μετοχών» (2018-10-10).**
Η μεγάλη ανοδική κίνηση που προηγείται και η μαξική πώληση αποτυπώνονται με σωστά αλλά με λίγη απαισιοδοξία ως προς το πλάτος. Στην απότομη πτώση έχουμε πλήρη χρονική αντιστοιχία.

**PG – «Ηρεμία Ανεπηρέαστη από Μακροοικονομικούς Παράγοντες» (2015-06-15).**
Σε περιβάλλον χαμηλού κινδύνου το μοντέλο επιτυγχάνει σχεδόν ισομετρική αντιστοίχιση. Οι αποκλίσεις είναι μικροσκοπικές και χωρίς συστηματικό πρόσημο.

### 12) Αποτελέσματα στρατιγικής - ποσοστιαία διαγράμματα

Ξεκινάμε από 0% κεφάλαιο και σε κάθε βήμα παίρνουμε θέση long όταν το μοντέλο προβλέπει άνοδο και θέση short όταν προβλέπει πτώση. Η καμπύλη κάθε μετοχής δεν είναι η τιμή της, αλλά το αποτέλεσμα του συγκεκριμένου κανόνα πάνω στη μετοχή. Αν η καμπύλη ανεβαίνει, σημαίνει ότι οι προβλέψεις έπιαναν σωστά το πρόσημο των κινήσεων εκείνη την περίοδο, αν κατεβαίνει, οι προβλέψεις ήταν, κατά μέσο όρο, αντίθετες από την πραγματική κίνηση. Οι απότομες βουτιές είναι drawdowns, ενώ ο τελικός ύψος της καμπύλης είναι η συνολική σωρευτική απόδοση.

Στο πρώτο διάγραμμα (Strategy Cumulative Returns) το μοντέλο αποδίδει καλύτερα σε TSLA, όπου η σωρευτική απόδοση κινείται για μεγάλο διάστημα πάνω από το 100% και φτάνει μέχρι ~150–170%, με εμφανείς αλλά διαχειρίσιμες υποχωρήσεις. Οι SPY και XOM διαμορφώνουν θετικές τροχιές γύρω στο 30–60%, υποδηλώνοντας αξιοπρεπή ικανότητα κατεύθυνσης στις ευρύτερες αγορές και στην ενέργεια. Η AMD παρουσιάζει βαθύ drawdown περίπου στο μέσο του δείγματος και στη συνέχεια ανάκαμψη που καταλήγει γύρω στο 60–80%. Η PG είναι πιο ήπια, με σταθερότερη πορεία κοντά στο 20–40%. Η JNJ καταλήγει χαμηλότερα (κοντά στο μηδέν), ενώ η AAPL κλείνει σχεδόν αρνητικά, κάτι που δείχνει ότι στις συγκεκριμένες περιόδους το μοντέλο δυσκολεύτηκε να πιάσει το πρόσημο των κινήσεων.

### 13) Αποτελέσματα στρατιγικής - διαγράμματα Sharp, Total Return, Max Drawdown

Τα τρία ραβδογράμματα, για κάθε μετοχή χωριστά, συνοψίζουν την απόδοση της ίδιας ακριβώς δοκιμαστικής στρατηγικής που τρέχουμε πάνω στις προβλέψεις του μοντέλου: 1) το Sharpe ratio μετρά πόση απόδοση ανά μονάδα κινδύνου πέτυχε η στρατηγική, 2) το Total return είναι το συνολικό κερδισμένο ποσοστό στο τέλος της περιόδου, ενώ 3) το Max drawdown είναι το μεγαλύτερο ποσοστό πτώσης της καμπύλης πλούτου της στρατηγικής, άρα δείχνει τον χειρότερο ενδιάμεσο «κραδασμό» που θα χρειαζόταν να αντέξουμε.

Οι μεγαλύτερες τελικές αποδόσεις έρχονται μαζί με μεγαλύτερα drawdowns. Η TSLA ξεχωρίζει με συνολικό κέρδος περίπου 120% και Sharpe γύρω στο 0.03, αλλά πληρώνεται με πολύ υψηλό μέγιστο drawdown λίγο περίπου 110%. Η AMD ακολουθεί με 59% total return και Sharpe 0.02, επίσης με βαριά ενδιάμεσα τραβήγματα (0.94). Ο SPY κινείται πιο ισορροπημένα με total return γύρω στο 44%, Sharpe κοντά στο 0.03 και αισθητά χαμηλότερο drawdown (0.31), κάτι που υποδηλώνει σταθερότερη συμπεριφορά του σήματος. Η XOM και η PG δίνουν μεσαίες επιδόσεις (περίπου 31% και 19% αντίστοιχα) με Sharpe 0.01–0.02 και drawdowns 0.70 και 0.30 περίπου. Η JNJ είναι οριακά θετική σε απόδοση (3%) και σχεδόν μηδενικό Sharpe, άρα η μετοχή εκεί δεν έχει ιδιάιτερες επιστροφές. Η AAPL είναι η αδύναμη εξαίρεση με αρνητικό τελικό αποτέλεσμα (18%) και αρνητικό  Sharpe, παρότι το drawdown παραμένει υψηλό. Στην πράξη, το μοντέλο δεν κατάφερε να εξάγει συνεπή κέρδη για αυτή τη μετοχή
.

### 14) Οπτικοποίηση τελικής βελτιστοποίησης σε σχέση με το αρχικό μοτέλο μηδενικής προσαρμογής

**1) Θερμικοί χάρτες σφαλμάτων**<br>
Ο πρώτος θερμικός χάρτης αποτυπώνει τη μείωση του σφάλματος του τελικού μοντέλου έναντι του μοντέλου μηδενικής προσαρμογής, ανά μετρική (MAE, RMSE, MAPE) και ανά μετοχή. Οι τιμές είναι απόλυτες διαφορές για το MAE/RMSE εκφράζονται σε μονάδες τιμής (δολάρια), ενώ για το MAPE σε ποσοστιαίες μονάδες. Θετικές τιμές σημαίνουν ότι το τελικό μοντέλο έχει μικρότερο σφάλμα από το μοντέλο μηδενικής προσαρμογής.

Η TSLA καταγράφει πάρα πολύ μεγάλη βελτίωση σε όλες τις μετρικές (≈112 στο MAE, ≈129 στο RMSE, ≈57 στο MAPE). Πρόκειται για μεγάλη υποχώρηση των μεγάλων λαθών (RMSE), ένδειξη ότι η τελική μορφή εκπαίδευσης ελέγχει πολύ καλύτερα την έντονη μεταβλητότητα της μετοχής. Ακολουθεί ο SPY με σημαντικές μειώσεις (≈31–34 σε MAE/RMSE και ≈8.6 σε MAPE), που δείχνουν σαφή βελτίωση. Οι AAPL και AMD εμφανίζουν ουσιώδεις, αλλά πιο μετριοπαθείς, μειώσεις σφάλματος, συμβατές με τα διαγράμματα απόδοσης. Αντίθετα, οι XOM, PG και ιδιαίτερα η JNJ παρουσιάζουν μικρότερη βελτίωση. Το τελικό μοντέλο κερδίζει, αλλά το περιθώριο ήταν εξαρχής περιορισμένο.


**2) Θερμικοί χάρτες κατευθυνσης**<br>
Ο δεύτερος θερμικός χάρτης δείχνει τη μεταβολή στις κατευθυντικές μετρικές σε απόλυτες μονάδες (0.01 ≈ 1 ποσοστιαία μονάδα). Η SPY ξεχωρίζει με την πιο συνεπή βελτίωση σε όλες τις μετρικές (\~+2.4 hit rate, +2.8 precision, +2.3 recall, +2.8 F1), γεγονός που ευθυγραμμίζεται με τη μεγάλη μείωση σφάλματος. Το μοντέλο προβλέπει συχνότερα τη σωστή κατεύθυνση και ταυτόχρονα παράγει πιο ευθυγραμμισμένα σήματα. Η TSLA εμφανίζει έντονη άνοδο στο recall (~ +3.3) με ουδέτερη/οριακά αρνητική μεταβολή στην precision. το μοντέλο γίνεται πιο ευαίσθητο στο να εντοπίζει κινήσεις, χωρίς όμως ανάλογη βελτίωση στη ορθότητα των θετικών σημάτων, κάτι. Η AAPL καταγράφει μικρές αλλά συνεπείς θετικές μεταβολές, ενώ η XOM έχει μικρά κέρδη, κυρίως στο recall/F1. Στις JNJ, AMD και PG οι διαφορές είναι οριακές ή ελαφρά αρνητικές σε hit rate/precision σε αυτά τα ήρεμα προφίλ το τελικό μοντέλο δεν αλλάζει ουσιαστικά την κατευθυντική συμπεριφορά, παρότι στις απόλυτες μετρικές σφάλματος υπήρξε βελτίωση.

**3) Wilcoxon**<br> 
Για τη σύγκριση FinalModel vs Zero-Shot, τα p-values είναι περίπου 0.0156 για MAE, RMSE, MAPE και Recall, 0.078 για F1, 0.687 για Precision και 0.938 για Hit-rate. Αυτό σημαίνει ότι, στο επίπεδο σημαντικότητας 5%, υπάρχει στατιστικά σημαντική διαφορά, ενώ για F1 η ένδειξη είναι οριακή (δεν απορρίπτεται στο 5%). Για Precision και Hit-rate δεν προκύπτει διαφορά (p πολύ μεγαλύτερα του 0.05) οπότε οι διαφορές είναι στατιστικά ασημαντές.

**Συμπέρασμα**<br>
Συνολικά, τα heatmaps επιβεβαιώνουν ποσοτικά ό,τι φάνηκε στα προηγούμενα αποτελέσματα. Η εκπαίδευση προσφέρει μεγάλη και συνεπή μείωση σφαλμάτων στις πιο δύσκολες χρονοσειρές, μεταφράζοντας μέρος αυτής της κερδισμένης ακρίβειας και σε κατευθυντική υπεροχή (ιδίως στον SPY). Σε μετοχές χαμηλότερης μεταβλητότητας, οι βελτιώσεις είναι μικρότερες και αφορούν κυρίως τις τιμολογιακές μετρικές, με τις κατευθυντικές διαφορές να παραμένουν κοντά στο μηδέν, κάτι που συνάδει με το περιορισμένο περιθώριο βελτίωσης σε σταθερά καθεστώτα μεταβλητότητας.

### 14) Συμπεράσματα 

Η τελική εκπαίδευση υπερέχει καθαρά του μοντέλου μηδενικής προσαρμογής. Η υπεροχή είναι εντονότερη στις πιο δύσκολες και ευμετάβλητες χρονοσειρές (ιδίως TSLA), όπου παρατηρείται μεγάλη μείωση σφάλματος (MAE/RMSE/MAPE) και, κατ’ επέκταση, ουσιαστική κατευθυντική βελτίωση. Στις πιο ήρεμες μετοχές (XOM, PG, JNJ) τα κέρδη είναι μεν υπαρκτά αλλά μικρότερα και επικεντρώνονται κυρίως στις απόλυτες μετρικές σφάλματος, με τις κατευθυντικές μεταβολές να παραμένουν κοντά στο μηδέν.

Τα αποτελέσματα του Wilcoxon ενισχύουν το παραπάνω. Οι βελτιώσεις σε MAE, RMSE, MAPE και Recall είναι στατιστικά σημαντικές στο 5%, το F1 είναι οριακό, ενώ Precision και Hit-rate δεν διαφοροποιούνται σημαντικά. Με πρακτικούς όρους, το τελικό μοντέλο συρρικνώνει τα μεγάλα λάθη και ανιχνεύει συχνότερα τις κινήσεις, χωρίς όμως να αυξάνει αναλόγως την ακρίβεια στις προβλέψεις θετικών σημάτων.

Η τελική βελτιστοποίηση αποδίδει ουσιαστικά οφέλη εκεί όπου έχει μεγαλύτερη σημασία, στην κλίμακα των λαθών και στον έλεγχο της μεταβλητότητας και μεταφράζει μέρος αυτής της ακρίβειας σε πιο συνεπή κατεύθυνση, ιδιαίτερα στον SPY.
