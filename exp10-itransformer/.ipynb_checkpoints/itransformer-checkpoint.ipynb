### 3) Οπτική αξιολόγηση iTransformer

Τα διαγράμματα του iTransformer δείχνουν πολύ σφιχτή επικάλυψη μεταξύ πρόβλεψης και πραγματικής τιμής στις περισσότερες μετοχές.

Στις AAPL, JNJ, PG η πρόβλεψη είναι σχεδόν επάνω στην τιμή, με πολύ μικρές αποκλίσεις γύρω από τα τοπικά άκρα.

Στις TSLA και AMD αναπαράγεται σωστά η μεσοπρόθεσμη δομή, αλλά τα ακραία άλματα αποτυπώνονται πιο ρηχά με τις κορυφές να είναι ελαφρά κατεσταλμένες.

Στην XOM έχουμε πολύ καλή στοίχιση στις ανοδικές και πτωτικές φάσεις του πετρελαϊκού κύκλου, αλλά μικρή χρονική υστέρηση σε απότομες διορθώσεις.

Στην SPY η ευρύτερη τάση συλλαμβάνεται καθαρά, αλλά ανά περιόδους η πρόβλεψη φαίνεται λίγο πιο λεία από την πραγματικότητα.

Το τελικό μοντέλο έχει λιγότερα πολύ μεγάλα σφάλματα και είναι πιο συμμετρικό αλλά έχει τιμλες που αποκλίνουν από το μηδέν οπότε περιμένουμε να δούμε λίγο μικρότερες μετρικές στα σφάλματα για τον iTransformer.

### 4) Αποτελέσματα μετρικών στο test set

* **Καλύτεροι (χαμηλότερα σφάλματα)**:
  **XOM** (MAE ≈ 0.93, RMSE ≈ 1.29), **PG** (1.07, 1.53), **JNJ** (1.10, 1.59).
  Πρόκειται για τίτλους με πιο ομαλή δυναμική· το μοντέλο τους «πιάνει» καλά.
* **Μεσαία επίδοση**: **AAPL** (1.72, 2.44), **AMD** (2.16, 3.20), **SPY** (2.93, 4.21).
  Στον SPY τα απόλυτα σφάλματα είναι υψηλότερα, αλλά **το MAPE είναι χαμηλό (≈ 0.79%)**, δηλ. το σχετικό σφάλμα ως προς το επίπεδο τιμής είναι μικρό.
* **Δύσκολη περίπτωση**: **TSLA** (MAE ≈ 6.23, RMSE ≈ 9.18, MAPE ≈ 3.21%).
  Η υψηλή βηματική μεταβλητότητα και οι απότομες αιχμές «κοστίζουν» αισθητά.

**Ανάγνωση της ράβδου MAPE.** Παρά τις υψηλές απόλυτες αποκλίσεις στον SPY, το χαμηλό MAPE δείχνει ότι, αναλογικά με το επίπεδο τιμών, το σφάλμα είναι συγκρατημένο. Αντίθετα, σε χαμηλότερου επιπέδου τίτλους, ένα μικρό απόλυτο λάθος μεταφράζεται σε μεγαλύτερο MAPE.


* **Υψηλότερη F1**: **JNJ** (≈ 0.535) και **SPY** (≈ 0.533).
  Ισορροπία precision–recall, δηλαδή καλή συνολικά κατηγοριοποίηση ανοδικών/πτωτικών ημερών.
* **Μεσαία ομάδα**: **XOM** (≈ 0.516), **PG** (≈ 0.520), **AAPL** (≈ 0.519).
  Οι τιμές βρίσκονται οριακά πάνω από 0.5, ένδειξη ελαφρού πλεονεκτήματος έναντι τυχαίας κατεύθυνσης.
* **Υστερούν**: **TSLA** (≈ 0.505), **AMD** (≈ 0.473).
  Η έντονη ασυμμετρία/μεταβλητότητα μειώνει κυρίως το **precision**, άρα υπάρχουν περισσότερα ψευδώς θετικά σήματα.

**Εύρος τιμών.** Το hit-rate κινείται \~0.48–0.53, με κορυφή το **JNJ** (≈ 0.533). Η μικρή διακύμανση ανάμεσα σε precision/recall ανά ticker σημαίνει ότι το μοντέλο δεν «θυσιάζει» συστηματικά το ένα για το άλλο, εκτός από τα πιο νευρικά ονόματα όπου το precision μειώνεται.




* **Αριστερό (MAE/RMSE/MAPE):** ιεραρχεί ξεκάθαρα τη δυσκολία πρόβλεψης ανά ticker. Όσο ψηλότερες οι ράβδοι, τόσο μεγαλύτερη η διασπορά των λαθών. Η **TSLA** ξεχωρίζει ως outlier, ενώ **XOM/PG/JNJ** αποτελούν «ασφαλές έδαφος» για τον iTransformer.
* **Δεξί (Hit-Rate/Precision/Recall/F1):** αποτυπώνει την ικανότητα κατεύθυνσης. **JNJ** και **SPY** παρέχουν τα πιο αξιόπιστα σήματα, **AMD/TSLA** τα πιο ασταθή. Η μικρή απόσταση ανάμεσα σε precision και recall υποδηλώνει ισορροπημένη συμπεριφορά (χωρίς έντονο bias προς over/under-signalling).

###  Αποτελέσματα μετρικών ανά καθεστώς μεταβλητότητας

Ο πίνακας συνοψίζει την απόδοση του iTransformer αφού ομαδοποιήσαμε τις περιόδους του test set σε τέσσερα καθεστώτα. Η εικόνα είναι συνεπής με τα ευρήματα στα «παράθυρα γεγονότων» και δείχνει σαφείς διαφορές ευρωστίας ανά περιβάλλον.

**High-Volatility.** Είναι το πλέον απαιτητικό καθεστώς: τα σφάλματα κορυφώνονται (**MAE≈4.20, RMSE≈6.19, MAPE≈2.79**) και οι κατευθυντικές μετρικές υποχωρούν (**Hit≈0.483, F1≈0.489**). Οι απότομες, ασύμμετρες μεταβολές και οι χοντρές ουρές δημιουργούν έντονο distribution shift, με αποτέλεσμα το μοντέλο να “εξομαλύνει” την κίνηση και να χάνει τόσο επίπεδο όσο και πρόσημο.

**Low-Volatility.** Πρόκειται για το «φυσικό» πεδίο του μοντέλου: εμφανίζει τα μικρότερα σφάλματα (**MAE≈1.08, RMSE≈1.56**) και το υψηλότερο κατευθυντικό σήμα (**Hit≈0.512, F1≈0.527**). Η ομαλή δυναμική και η μεγαλύτερη προβλεψιμότητα των ροών τιμών ευνοούν την προσοχή/μάθηση του Transformer.

**Post-Trend-Reversal.** Οι επιδόσεις παραμένουν καλές και ισορροπημένες (**MAE≈1.33, RMSE≈1.86, F1≈0.517**). Οι μεταβάσεις από down σε up trend «διαβάζονται» επαρκώς, αν και η σχετική ακρίβεια (MAPE≈1.37) είναι χειρότερη από το low-vol, στοιχείο που υποδηλώνει μέτρια δυσκολία στην ταχύτητα προσαρμογής αμέσως μετά το γύρισμα.

**Sideways/Chop.** Παρότι το **MAPE** είναι το χαμηλότερο (**≈0.80**), τα απόλυτα σφάλματα είναι αυξημένα (**MAE≈2.93, RMSE≈4.21**)—ένδειξη ότι οι τιμές βρίσκονται σε υψηλά επίπεδα (συνεπώς το απόλυτο λάθος μεγαλώνει) ενώ το μοντέλο συλλαμβάνει σχετικά καλά την αναλογική μεταβολή. Κατευθυντικά κινείται αξιοπρεπώς (**F1≈0.533**) παρά τον «θόρυβο» προσήμων.

**Ερμηνεία και συνέπειες.**

1. Το μοντέλο είναι **ισχυρό σε low-vol και post-reversal** περιβάλλοντα∙ εκεί μπορεί να χρησιμοποιηθεί πιο «επιθετικά».
2. Σε **high-vol** καθεστώς συνιστάται **μείωση μόχλευσης/θέσης** ή πιο αυστηρά thresholds για σήμα κατεύθυνσης (το hit-rate <0.49 υπονοεί ότι ένα «άκριτο» long/short θα υποφέρει).
3. Στο **sideways**, η καλή σχετική ακρίβεια αλλά υψηλό MAE υποδεικνύει χρήσιμη κατευθυντική πληροφόρηση για βραχείες κινήσεις, αλλά με **προσοχή στο risk budget** λόγω απόλυτου λάθους.

**Τεχνικές βελτίωσης που προκύπτουν από τον πίνακα.**
– **Regime-aware εκπαίδευση** (features/conditioning με realized vol, VIX, drawdown state).
– **Robust losses** (Huber/quantile) ή heavy-tail likelihoods για high-vol.
– **Διαφορετικά heads/weights ανά καθεστώς** ή mixture-of-experts.
– **Κανόνες εκτέλεσης** ανά καθεστώς (π.χ. scaling θέσης με vol, dynamic thresholds για buy/sell).

Συνολικά, ο iTransformer παρουσιάζει **σαφή πλεονεκτήματα σε ήπια και μεταβατικά καθεστώτα** και **αδυναμία σε επεισοδιακή έντονη μεταβλητότητα**—ακριβώς το προφίλ που θα περιμέναμε από ένα sequence model που έχει εκπαιδευτεί σε MSE-τύπου στόχο χωρίς ρητή μοντελοποίηση ακραίων ουρών.


$$
\begin{array}{lccccccc}

\hline

Low-Volatility & 1.081556 & 1.563383 & 0.813544 & 0.511576 & 0.520468 & 0.534619 & 0.527448 \\

### 6) Διαγράμματα κυλιόμενης μεταβλητότητας - σφάλματος

Τα διαγράμματα διασποράς απεικονίζουν την απόλυτη τιμή του σφάλματος πρόβλεψης σε συνάρτηση με τη 5-ημερη κυλιόμενη μεταβλητότητα (log-return std). Και στα επτά tickers παρατηρείται σαφής ετεροσκεδαστικότητα: όσο αυξάνει η βραχυπρόθεσμη μεταβλητότητα τόσο «ανοίγει» ο κώνος των σφαλμάτων. Το κρίσιμο εύρημα είναι ότι το νέφος του **iTransformer** (μπλε) πλαταίνει πολύ πιο έντονα από αυτό του **FinalModel** (πορτοκαλί), ιδίως στα ανώτερα δεκατημόρια της μεταβλητότητας. Σε χαμηλή μεταβλητότητα τα δύο μοντέλα συγκλίνουν· η απόκλιση εμφανίζεται και επιτείνεται μόλις το std υπερβεί \~2–3% (σε όρους log-returns).

Ανά ticker, η εικόνα είναι συνεκτική. Στην **AAPL** η σχέση μεταβλητότητας–σφάλματος είναι εμφανώς ανοδική, με τον iTransformer να παράγει περισσότερα μεγάλα λάθη για vol >0.05, ενώ το FinalModel παραμένει συμπαγές. Η **TSLA** παρουσιάζει το ισχυρότερο φαινόμενο: στην υψηλή μεταβλητότητα συσσωρεύονται μπλε outliers σε επίπεδα σφάλματος 20–50, όταν το πορτοκαλί νέφος σπανίως ξεπερνά τα \~15. Στην **XOM** και στο **SPY** το πρότυπο είναι ηπιότερο αλλά ίδιο: ήπια κλίση σε χαμηλό vol και αισθητή διεύρυνση των λαθών του iTransformer καθώς το vol κινείται προς 0.04–0.08. Η **JNJ** και η **PG**, ως αμυντικές, έχουν συνολικά μικρότερες διασπορές και σχετικά αδύναμη κλίση∙ ωστόσο ακόμη και εκεί, τα λίγα επεισόδια αυξημένου vol συνοδεύονται από δυσανάλογες εκρήξεις σφάλματος στον iTransformer. Η **AMD** προσομοιάζει την TSLA: ισχυρή κυρτότητα του νέφους και σαφώς περισσότερα μεγάλα μπλε λάθη όταν το vol περνά το 0.05–0.06.

Η συνολική ανάγνωση είναι ότι ο iTransformer εμφανίζει **υπερευαισθησία στη μεταβλητότητα**: η αύξηση των σφαλμάτων με το vol είναι ισχυρότερη, με περισσότερα ακραία errors και παχύτερες ουρές σε σχέση με το FinalModel. Αυτό συνάδει με τα ευρήματα καθεστώτων (υποχώρηση σε High-Vol) και υποδεικνύει ότι το μοντέλο δεν προσαρμόζει επαρκώς τη δυναμική του όταν η διαδικασία μετάβασης γίνεται πιο θυελλώδης.

Πρακτικά, τα διαγράμματα υποστηρίζουν τρεις κατευθύνσεις βελτίωσης: πρώτον, ρητή **conditioned** μοντελοποίηση στη μεταβλητότητα (features όπως realized vol/VIX ή regime tokens) ώστε η προσοχή να αλλάζει λειτουργία ανά περιβάλλον. Δεύτερον, **robust** στόχοι και πιθανότητες (Huber/quantile ή heavy-tailed likelihoods) για τον περιορισμό των ακραίων σφαλμάτων. Τρίτον, **vol-scaling** των τιμών/returns στη μάθηση και κανόνες διαχείρισης θέσης που συρρικνώνουν την έκθεση όταν το vol ανεβαίνει. Με αυτή τη στόχευση, αναμένουμε το νέφος του iTransformer να συγκλίνει προς το πιο «σφιχτό» πορτοκαλί μοτίβο του FinalModel, ιδίως στο πάνω τμήμα της κατανομής μεταβλητότητας.

### Pearson συσχέτιση (μεταβλητότητα ↔ σφάλμα) για iTransformer — ερμηνεία

Οι τιμές είναι ο συντελεστής Pearson μεταξύ της 5-ημερης κυλιόμενης μεταβλητότητας (std των log-returns) και του απόλυτου σφάλματος πρόβλεψης. Θετικές τιμές σημαίνουν ότι, όταν αυξάνει η βραχυπρόθεσμη μεταβλητότητα, τείνει να αυξάνει και το μέγεθος του σφάλματος.

**Επίπεδο συσχέτισης.**
Όλα τα tickers έχουν **θετική** συσχέτιση∙ τα περισσότερα κινούνται στη ζώνη **0.20–0.38** (μέτρια συσχέτιση). Η **PG** ξεχωρίζει με **0.5654** (ισχυρή συσχέτιση), ακολουθούν **AAPL ≈ 0.38** και **XOM ≈ 0.34**, ενώ οι χαμηλότερες τιμές εμφανίζονται σε **TSLA (0.21)** και **AMD (0.22)**.

**Σειρά ευαισθησίας (φθίνουσα):** PG » AAPL ≳ XOM > SPY ≳ JNJ ≳ AMD ≳ TSLA.
Η κατάταξη αυτή συνάδει με τα scatter plots: στις οξείες φάσεις μεταβλητότητας, τα σφάλματα της PG (και, σε μικρότερο βαθμό, της AAPL/XOM) διογκώνονται αισθητά.

**Σε σχέση με άλλα μοντέλα.**

* Έναντι **FinalModel**: ο iTransformer δείχνει **εντονότερη** σύζευξη με τη μεταβλητότητα σε **AAPL, AMD, PG** (μεγάλη διαφορά στην PG), αλλά **ηπιότερη** σε **XOM, SPY, JNJ**.
* Έναντι **ARIMA**: γενικά **ασθενέστερη** σύζευξη (εκτός της PG, όπου παραμένει υψηλή).
* Έναντι **LSTM**: το LSTM εμφάνιζε μηδενικές/αρνητικές συσχετίσεις∙ ο iTransformer είναι **σαφώς πιο προβλέψιμα «ευαίσθητος»** στη μεταβλητότητα.

**Τι σημαίνει πρακτικά.**
Ο iTransformer, συνολικά, διατηρεί **μέτρια** εξάρτηση σφάλματος από το βραχυπρόθεσμο risk, με **«ευαίσθητα» σημεία** σε PG και AAPL. Σε περιόδους υψηλού vol αναμένονται μεγαλύτερες αποκλίσεις (ιδίως για PG), άρα στη χρήση του μοντέλου είναι χρήσιμα:

* vol-aware κανονικοποίηση/κλιμάκωση και **regime tokens** ώστε η προσοχή να «γυρίζει» προφίλ σε stress φάσεις,
* **robust** loss (Huber/quantile) για περιορισμό ακραίων λαθών,
* και **διαχείριση θέσης** που συρρικνώνει την έκθεση όταν ο βραχυπρόθεσμος std ανεβαίνει.

Συμπέρασμα: ο iTransformer βρίσκεται «στη μέση» μεταξύ ARIMA (πολύ ευαίσθητο) και LSTM (υπο-ευαίσθητο), με στοχευτές βελτιώσεις σε vol-aware μάθηση να μπορούν να μειώσουν κυρίως την ισχυρή σύζευξη που βλέπουμε στην PG και, δευτερευόντως, στην AAPL.

### Αποτελέσματα στα παράθυρα γεγονότων

**Κατάρρευση & Ανάκαμψη (AAPL, 2020-03-10).** Η ανοδική φάση πριν το σοκ αποδίδεται πιστά, το απότομο ξεπούλημα αναπαράγεται με μεγάλη  καθυστέρηση η οποία στη φάση της ανάκαμψης ακολουθείται μειώνεται.

**TSLA – High-Beta Cooling (Ιαν. 2021).** Το ανοδικό σκέλος αποτυπώνεται σωστά, με ήπια υποεκτίμηση. Η αποφόρτιση μετά την κορυφή συλλαμβάνεται επίσης σωστά, αλλά υπάρχει καθυστέρηση κάποιων ημερών στην τιμή και γενικά τα μικρά σκαμπανεβάσματα δεν αποτυπώνονται σωστά.

**XOM – Oil Cycle Peak (Ιούν. 2022).** Η κλιμακωτή άνοδος του κύκλου πετρελαίου παρακολουθείται πάρα πολύ καλά. Το σημείο κορύφωσης/στροφής καταγράφεται με μικρή καθυστέρηση και τα επόμενα κυματικά μοτίβα αναπαράγονται με καλή χρονική στοίχιση. Είναι η καλύτερη απόδοση για το παράθυρο της XOM που έχουμε δει μέχρι τώρα.

**SPY – Drawdown Chop (Φθινόπωρο 2022).** Αν και η γενική κατεύθυνση είναι σωστή ο iTransformer έχει κάνει πάρα πολυ μεγάλη εξομάλυνση της τιμής, και λόγω αυτού θα είναι λογικό οι κατευθυντικές μετρικές να είναι σε πολύ χαμηλά επίπεδα.

**JNJ – Low-Volatility Stretch (2019).** Σε περιβάλλον χαμηλής μεταβλητότητας η πρόβλεψη κάθεται πρακτικά πάνω στην τιμή, με αμελητέες αποκλίσεις ακόμη και γύρω από βραχείες ανωμαλίες.

**AMD – Tech Selloff (Οκτ. 2018).** Αρκετά παρόμοιο αποτέλεσμα με αυτό της TSLA.

**PG – Macro-Irrelevant Calm (2015).** Εξαιρετικά στενό ταίριασμα σε όλο το παράθυρο.

### iTransformer — απόδοση σε «γεγονότα» (event windows)

Οι μετρικές του πίνακα αποτυπώνουν την επίδοση του μοντέλου σε σύντομα και «ιδιόμορφα» υποσύνολα της αγοράς (crash/rebound, peaks, sell-off, κ.λπ.). Επειδή τα παράθυρα είναι μικρά, οι διαφορές έναντι του συνολικού test set γίνονται εντονότερες και η διακύμανση των μετρικών αυξάνει.

Στο επίπεδο σφάλματος, ο iTransformer βελτιώνεται καθαρά σε **AAPL, JNJ, AMD, PG** (λ.χ. AAPL: MAE ≈ 1.14 από \~1.72, JNJ: \~0.86 από \~1.10, PG: \~0.42 από \~1.07), ενώ χειροτερεύει σε **SPY, TSLA, XOM** (SPY: MAE \~4.57 από \~2.93· TSLA: \~6.40 από \~6.23· XOM: \~1.20 από \~0.93). Η εικόνα του **MAPE** βοηθά να ερμηνευτούν οι «αντιφάσεις»: για την **AMD** το MAE είναι πολύ χαμηλό επειδή το επίπεδο τιμών στο sell-off του 2018 ήταν μικρό, αλλά **το MAPE ανεβαίνει (\~2.99%)**, υποδηλώνοντας ότι αναλογικά το λάθος παραμένει αισθητό. Αντίθετα, στον **SPY** το MAPE αυξάνεται απότομα (\~1.17% από \~0.79%), στοιχείο που δείχνει ότι ο τεμαχισμένος «drawdown-chop» κάνει δύσκολη και τη σχετική, όχι μόνο την απόλυτη, πρόβλεψη.

Ως προς την κατεύθυνση, ο iTransformer αποδίδει καλύτερα σε περιβάλλοντα με πιο καθαρή δυναμική. **XOM** και **JNJ** καταγράφουν την ισχυρότερη συνολική ταξινόμηση (**F1 ≈ 0.57 και 0.55**, με ταυτόχρονη άνοδο τόσο του precision όσο και του recall έναντι του test set), ενώ ο **PG** παραμένει σταθερός γύρω από **hit-rate ≈ 0.51**. Στον αντίποδα, ο **SPY** στο «Drawdown-Chop» εμφανίζει σημαντική πτώση κατευθυντικών μετρικών (**precision ≈ 0.41, F1 ≈ 0.41**), γεγονός συνεπές με το αποσπασματικό, εναλλασσόμενο μοτίβο των ημερήσιων επιστροφών. Η **TSLA** παραμένει απαιτητική: τα σφάλματα είναι υψηλά και η F1 κινείται οριακά πάνω από 0.45, κάτι που αντικατοπτρίζει την υψηλή βήτα και τις μη γραμμικές αιχμές. Ο **AAPL** βελτιώνει έντονα τα σφάλματα στη φάση «Crash & Rebound», αλλά οι κατευθυντικές μετρικές μένουν γύρω στο 0.47, δείχνοντας ότι η μεγάλη ανάκαμψη «πιάνεται» ως επίπεδο, όχι πάντα ως σωστή ημερήσια πρόσημη.

Συνοπτικά, στα παράθυρα γεγονότων ο iTransformer:

* **Κρατά ή βελτιώνει** την ακρίβεια επιπέδου όταν η δυναμική είναι πιο ομαλή ή κυκλική (**JNJ, XOM, PG** για επίπεδο· **JNJ, XOM** και κατευθυντικά).
* **Υστερεί** σε **choppy** περιβάλλοντα και τίτλους υψηλής αστάθειας (**SPY drawdown-chop, TSLA**), όπου πέφτει κυρίως το precision.
* **Χρειάζεται προσοχή** στην ερμηνεία MAE/MAPE: σε χαμηλά επίπεδα τιμών (π.χ. AMD 2018) μικρά απόλυτα λάθη μπορεί να αντιστοιχούν σε ουσιαστικά σχετική απόκλιση.

Πρακτική συνέπεια: το μοντέλο είναι καταλληλότερο για γεγονότα με **καθαρή κατεύθυνση/τάση** και για **αμυντικούς** ή **κυκλικούς** τίτλους, ενώ σε **chop** ή **υψηλή βήτα** απαιτείται ενίσχυση με σήματα καθεστώτος (volatility regime features) ή/και loss που τιμωρεί ασύμμετρα τα λάθη σε καταιγιστικές κινήσεις.

Το γράφημα απεικονίζει την **εξέλιξη του σωρευτικού κέρδους** της στρατηγικής (σε ποσοστό) για κάθε μετοχή στο χρόνο. Ο οριζόντιος άξονας είναι διακριτά χρονικά βήματα του test set, ενώ ο κατακόρυφος δείχνει το συνολικό αποτέλεσμα από την αρχή (0% σημαίνει «ούτε κέρδος ούτε ζημία»). Κάθε καμπύλη αντιστοιχεί σε ένα ticker. Η εικόνα δείχνει καθαρή **ετερογένεια ανά μέσο**: η **TSLA** (πορτοκαλί) κυριαρχεί με μεγάλα ανοδικά σκέλη και έντονα retracements, αλλά κλείνει υψηλά (\~110–130%), στοιχείο υψηλής απόδοσης με υψηλή μεταβλητότητα. Η **AMD** (καφέ) παρουσιάζει επίσης ισχυρές εκρηκτικές φάσεις (κορύφωση >100%) με βαθιά ενδιάμεσα drawdowns. Ο **SPY** (κόκκινο) ακολουθεί ηπιότερη, πιο «ομαλή» ανιούσα τροχιά που τελικά σταθεροποιείται γύρω στο 60–70%, υποδηλώνοντας καλύτερο λόγο απόδοσης/κινδύνου. Το **PG** (ροζ) κινείται σταθερά με μικρά εύρη και μέτριο τελικό κέρδος (\~20–30%), ενώ το **JNJ** (μωβ) και το **XOM** (πράσινο) καταγράφουν αρχικά αξιοπρόσεκτα κέρδη που αμβλύνονται στο δεύτερο μισό της περιόδου. Η **AAPL** (μπλε) είναι η πιο αδύναμη καμπύλη, με παρατεταμένες φάσεις υποχώρησης και αρνητικό ή οριακά θετικό κλείσιμο. Συνολικά, οι καμπύλες καταδεικνύουν ότι το μοντέλο αποδίδει καλύτερα όπου επικρατούν **σαφείς και επίμονες τάσεις** (SPY, TSLA, AMD), ενώ σε πιο «κυκλικά» ή mean-reverting περιβάλλοντα η απόδοση διαβρώνεται, αναδεικνύοντας την ανάγκη για **διαχείριση μεταβλητότητας** (volatility targeting) ώστε να περιορίζονται τα μεγάλα ενδιάμεσα drawdowns των high-beta ονομάτων.

### Αξιολόγηση οικονομικών με iTransformer (Sharpe – Cumulative Return – Max Drawdown)

**Τι δείχνουν τα διαγράμματα.**
Τα τρία ραβδογράμματα συνοψίζουν την πορεία μιας απλής long-only στρατηγικής που ανοίγει θέση σύμφωνα με το πρόσημο της εκτιμώμενης (log) απόδοσης του iTransformer. Παρουσιάζονται: (α) ο **Sharpe ratio** (ετήσια προσαρμογή, risk-free≈0), δηλαδή απόδοση ανά μονάδα διακύμανσης∙ (β) η **συνολική σωρευτική απόδοση** της στρατηγικής (equity\_T − 1)∙ και (γ) το **μέγιστο drawdown**, δηλαδή η χειρότερη πτώση του equity από το προηγούμενο υψηλό του (όσο μεγαλύτερη η μπάρα, τόσο πιο οδυνηρή η πτώση).

**Κύρια ευρήματα ανά δείκτη.**

* **Sharpe ratio.** Η καλύτερη αναλογία απόδοσης/κινδύνου εμφανίζεται στο **SPY** (≈0.04), ένδειξη ότι οι προβλέψεις του iTransformer μεταφράζονται σε σχετικά «καθαρές» κινήσεις στο ευρύτερο market. Ακολουθούν **AMD** και **TSLA** (≈0.02–0.025), έπειτα **PG** (\~0.01), ενώ **XOM** και **JNJ** κινούνται οριακά θετικά. Η **AAPL** είναι ουσιαστικά μηδενική/ελαφρά αρνητική. Συνεπώς, από πλευράς **risk-adjusted** απόδοσης το μοντέλο λειτουργεί πιο αξιόπιστα σε index/mega-caps (SPY) και σε ορισμένα high-beta (AMD/TSLA) όταν γίνει σωστή διαχείριση κινδύνου.

* **Cumulative return.** Οι υψηλότερες σωρευτικές αποδόσεις προκύπτουν σε **TSLA** (\~112%) και **AMD** (\~111%), ακολουθεί το **SPY** (\~71%). **PG** (\~22%) και **XOM** (\~11%) δίνουν μέτριες αποδόσεις, το **JNJ** χαμηλή (~~6%), ενώ η **AAPL** είναι αρνητική (~~−3%). Το μοτίβο υποδεικνύει ότι ο iTransformer «κεφαλαιοποιεί» καλά τις παρατεταμένες τάσεις σε high-beta ονόματα και στον ευρύτερο δείκτη, αλλά δυσκολεύεται σε πιο «αμυντικά» ή mean-reverting περιβάλλοντα.

* **Max drawdown.** Το κόστος ρίσκου είναι έντονο σε **TSLA** και **AMD** (πολύ υψηλά drawdowns), αυξημένο και στο **XOM**, πιο ήπιο στο **SPY** (η μικρότερη μπάρα μεταξύ των υψηλών αποδόσεων), χαμηλό στο **PG** και μέτριο στο **JNJ**. Με απλά λόγια, εκεί όπου κερδίζουμε πολλά (TSLA/AMD) πληρώνουμε και μεγάλα «βουναρά» ζημιών αν δεν υπάρχει έλεγχος θέσης.

**Συνδυαστική ερμηνεία (risk–return trade-off).**

* **Καλύτερο risk-adjusted προφίλ:** **SPY** (υψηλός Sharpe, ικανοποιητική απόδοση, σχετικά μικρό drawdown).
* **High-beta «μοχλευμένα» κέρδη με βαρύ ρίσκο:** **TSLA**, **AMD** (πολύ καλές σωρευτικές αποδόσεις, αλλά μεγάλα drawdowns· χρειάζεται αυστηρή διαχείριση μεταβλητότητας).
* **Αμυντικά/σταθερά:** **PG** (χαμηλά drawdowns, αλλά μικρό upside) και **JNJ** (χαμηλό return, ουδέτερος Sharpe).
* **Υπο-απόδοση:** **AAPL** (αρνητικό total return, μηδενικός Sharpe)∙ **XOM** χαμηλό return με σχετικά υψηλό drawdown.

**Πρακτικές συνέπειες.**
Για να εκμεταλλευτεί κανείς τα δυνατά σημεία του iTransformer χωρίς να εκτεθεί στις οξείες πτώσεις:

1. **Volatility-targeting / position sizing:** κλιμάκωση θέσης με βάση τον 5-ημερο std των log-returns ή το εμπιστοσύνολο της πρόβλεψης∙ μικρότερη έκθεση σε περιόδους υψηλού vol (ιδίως σε TSLA/AMD).
2. **Regime awareness:** ενσωμάτωση regime tokens ή φίλτρων (π.χ. trend vs chop) — τα αποτελέσματα δείχνουν ότι σε ευρύτερα, «καθαρά» trends (SPY) το μοντέλο αποδίδει καλύτερα.
3. **Robust losses / risk controls:** Huber/quantile loss και απλοί κανόνες προστασίας (trailing stops ή cap σε ημερήσιο draw) μειώνουν τις ουρές ζημιάς.
4. **Κόστη/τριβές:** τα νούμερα είναι προ-transaction costs· σε υλοποίηση απαιτείται έλεγχος turnover.

Συνολικά, ο iTransformer δείχνει **σταθερή υπεροχή σε SPY** και **δυνατότητα υψηλών κερδών σε TSLA/AMD** με αντάλλαγμα μεγαλύτερα drawdowns. Για «core» έκθεση προκρίνεται ο SPY με volatility-targeting, ενώ για «satellite» τοποθετήσεις μπορούν να αξιοποιηθούν TSLA/AMD υπό αυστηρό risk management.

### Heatmaps – τι δείχνουν και τι συμπεραίνουμε

**1) Error Improvement (iTransformer − FinalModel)**
Το heatmap απεικονίζει τη διαφορά σφάλματος σε *MAE/RMSE/MAPE* μεταξύ iTransformer και FinalModel. Οι τιμές είναι **(iTransformer − FinalModel)**, άρα **αρνητικά** νούμερα σημαίνουν ότι το FinalModel έχει **μικρότερο σφάλμα** (δηλ. είναι καλύτερο), και όσο πιο αρνητικό τόσο μεγαλύτερο το προβάδισμά του.

* Η υπεροχή του FinalModel είναι **καθολική**: όλα τα κελιά είναι αρνητικά.
* **Μεγαλύτερες αποστάσεις**:

  * **TSLA** (−0.72 MAE, −0.95 RMSE, −0.73 MAPE) και **SPY** (−0.99, −1.19, −0.25) — το iTransformer δυσκολεύεται στα πιο «νευρικά» ή πολυκαθεστωτικά σήματα.
  * **AMD** (−0.51, −0.63, −0.54) και **XOM** (−0.34, −0.36, −0.67) επίσης δείχνουν καθαρό πλεονέκτημα του FinalModel.
* **Μικρότερες αποστάσεις**: **PG** και **JNJ** (|Δ|≲0.25 στα MAE/RMSE) — σε αμυντικές/χαμηλής μεταβλητότητας μετοχές το iTransformer πλησιάζει, αλλά παραμένει πίσω.

**Διά ταύτα**: ως προς *απόλυτα σφάλματα τιμής*, το FinalModel προηγείται ξεκάθαρα σε όλα τα tickers, με το κενό να ανοίγει όσο αυξάνει η μεταβλητότητα/αλλαγή καθεστώτος.

---

**2) Directional Delta (FinalModel − iTransformer)**
Εδώ βλέπουμε τη διαφορά σε *hit\_rate/precision/recall/F1* (**Final − iTrans**). Θετικό = το FinalModel έχει υψηλότερη τιμή.

* **Συστηματικές θετικές βελτιώσεις** σε:

  * **TSLA** *(+0.010, +0.009, +0.019, +0.014)*,
  * **SPY** *(\~0, +0.000, +0.022, +0.010)*,
  * **AMD** *(+0.012, +0.012, +0.025, +0.018)*.
    Ιδίως στο **recall** (σωστή σύλληψη κατεύθυνσης), το FinalModel κερδίζει καθαρά.
* **Οριακά/μικτά** αποτελέσματα:

  * **AAPL** μικρές απώλειες σε hit/precision (\~−0.01) με ουδέτερο–ελαφρά θετικό recall, συνολικά ουδέτερο F1.
  * **PG** μικρές αρνητικές σε hit/precision (\~−0.009), μικρό + στο recall.
* **Μοναδική εξαίρεση**: **XOM**, όπου τα deltas είναι **αρνητικά** σε όλα τα directional metrics (περίπου −0.02). Αυτό σημαίνει ότι, παρότι το FinalModel έχει μικρότερα *σφάλματα τιμής* στο XOM (βλ. πρώτο heatmap), το iTransformer τείνει να «μαντεύει» ελαφρώς καλύτερα την **κατεύθυνση** στο ημερήσιο βήμα.

**Συνοπτική εικόνα**:

* Σε **σφάλμα τιμής** το FinalModel υπερέχει παντού.
* Σε **κατεύθυνση**, η υπεροχή του FinalModel είναι γενικά θετική και πιο έντονη σε **TSLA/SPY/AMD**, ουδέτερη σε **AAPL/PG**, ενώ στο **XOM** το iTransformer έχει ένα μικρό προβάδισμα.

---

**Πώς δένει με τα προηγούμενα ευρήματα;**

* Τα μεγαλύτερα κενά στα errors (TSLA, SPY, AMD) ταιριάζουν με όσα είδαμε σε scatter/volatility plots και Pearson: εκεί που η μεταβλητότητα/καθεστωτικές αλλαγές είναι υψηλές, το FinalModel «κρατά» καλύτερα.
* Τα μικρά και ανάμικτα deltas σε χαμηλού θορύβου tickers (PG, JNJ) εξηγούν γιατί στο Wilcoxon τα directional p-values δεν είναι παντού ισχυρά: οι διαφορές υπάρχουν, αλλά είναι **μικρής τάξης μεγέθους**.

**Κύριο συμπέρασμα**: Το FinalModel είναι σταθερά καλύτερο στα *errors* και, στις περισσότερες μετοχές, λίγο έως αισθητά καλύτερο στην *κατεύθυνση*. Το iTransformer πλησιάζει σε σταθερότερα ονόματα και κρατά ένα οριακό πλεονέκτημα direction στο XOM, αλλά συνολικά υστερεί.
