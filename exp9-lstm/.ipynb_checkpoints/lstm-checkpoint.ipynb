### 1) Περιγραφή πειράματος
Το LSTM (Long Short-Term Memory) είναι μια παραλλαγή αναδρομικών νευρωνικών δικτύων (RNN) σχεδιασμένη για να θυμάται πληροφορία σε μεγάλους χρονικούς ορίζοντες και να αποφεύγει το πρόβλημα του σβησίματος κλίσεων. Είναι κατάλληλο για χρηματιστηριακές χρονοσειρές επειδή μπορεί να αποτυπώσει μη γραμμικές σχέσεις, αλλαγές καθεστώτων και καθυστερημένες επιδράσεις.

Βασικό μοντέλο LSTM(βλ. θεωριτικό υπόβαθρο):<br>
- forget gate: αποφασίζει τι μέρος της παλιάς μνήμης (cell state) θα διαγραφεί.
- input gate: ελέγχει ποια νέα πληροφορία θα εισαχθεί στη μνήμη.
- candidate state: προτεινόμενη νέα πληροφορία που μπορεί να προστεθεί στη μνήμη.
- output gate: καθορίζει ποιο κομμάτι της μνήμης θα «εκτεθεί» ως κρυφή κατάσταση (hidden state) και άρα θα επηρεάσει την πρόβλεψη.
Ο συνδυασμός αυτών των πυλών επιτρέπει στο μοντέλο να διατηρεί/ξεχνά πληροφορία δυναμικά ανάλογα με το μοτίβο της χρονοσειράς.


#### Ερευνητικά Ερωτήματα Πειράματος 8 – LSTM

- Μια στρατηγικη σχεδιασμένη με LSTM μπορεί να αποφέρει περισσότερα κέρδη

### 2) Περιγραφή Μοντέλου & Εκπαίδευσης

**Στόχος πρόβλεψης**<br>
Η εκπαιδευση και οι προβλέψεις πρέπει να γίνουν σε ως ένα βαθμό στάσιμη χρονοσειρά (να αφαιρεθεί ο θόρυβος). Το μοντέλο προβλέπει τον επόμενης ημέρας λογαριθμικό ημερήσιο αριθμοδεικτή απόδοσης (log-return) της τιμής κλεισίματος:

$$
r_t = \log(P_t) - \log(P_{t-1})
$$

Η πρόβλεψη γίνεται σε επίπεδο return (όχι τιμής), και αργότερα μετατρέπεται σε τιμή. Με αυτόν τον τρόπο εξασφαλίζουμε κάποιο βαθμό στασιμότητας 

<br>

**Είσοδοι (features) & στόχος**<br>
Για κάθε μετοχή χρησιμοποιούνται οι ακόλουθες είσοδοι ανά ημέρα:

* open, high, low
* τεχνικοί δείκτες: sma_20, ema_20, wma_20, tema_20, bb_upper, bb_lower

Ο στόχος για εκπαίδευση είναι το $r_t$ που προκύπτει από τις τιμές close.

<br>

**Προεπεξεργασία & Κλίμακα**<br>
Υπολογισμός returns: από το close παράγονται τα $r_t$ (με np.diff(log_prices) και εισαγωγή $r_0=0$).<br>
Διαίρεση σε train/test: το **42%** των παρατηρήσεων χρησιμοποιείται ως train set (για να έχουμε ακριβή σύγκριση με το ΤΤΜ).<br>
Κλιμάκωση εισόδων: MinMaxScaler πάνω μόνο στο train set των χαρακτηριστικών εισόδου. Η μετασχηματισμένη κλίμακα εφαρμόζεται έπειτα στο σύνολο.<br>
Κλιμάκωση στόχου: StandardScaler πάνω μόνο στα returns της εκπαίδευσης. Το μοντέλο μαθαίνει σε scaled returns.

<br>

**Δημιουργία δειγμάτων (sliding windows)** <br>
Χρησιμοποιείται παράθυρο σταθερού μήκους LOOKBACK = 256 ημερών(το 512 δεν είχε καμμία διαφορά στο αποτέλεσμα).

Για κάθε χρονικό δείκτη $i$ στο training, το input είναι

  $$
  X_{i} \in \mathbb{R}^{256 \times F},\quad X_i = [x_{i-256}, \dots, x_{i-1}]
  $$

  όπου $F$ = αριθμός χαρακτηριστικών.
Ο στόχος για το ίδιο δείγμα είναι ο scaled return $r_i$. Για να υπάρχει $P_{i-1}$ στον υπολογισμό του return, τα έγκυρα indices ξεκινούν από lookback + 1.

<br>

**Αρχιτεκτονική LSTM**<br>
Το δίκτυο είναι σειριακό (Keras Sequential) με δύο LSTM στοιβαγμένα και ένα dense output:

1. Input(shape=(256, F))
2. LSTM(128, return_sequences=True)
3. LSTM(64)
4. Dense(1)

προβλέπει ένα scalar, τον επόμενης ημέρας scaled log-return.

<br>

**Loss & Optimizer**<br>
Loss: MSE πάνω στα scaled returns.<br>
Optimizer: Adam.

<br>

**Διαδικασία Εκπαίδευσης**<br>
Διαχωρισμός validation: Από τα training windows κρατείται το τελευταίο 10% ως validation set.<br>
Early Stopping: Παρακολουθείται val_loss με patience=6 και restore_best_weights=True, ώστε να αποφευχθεί υπερεκπαίδευση και να ανακτηθούν τα καλύτερα βάρη.

<br>
  
**Hyperparameters:** <br>
LOOKBACK = 256<br>
EPOCHS = 40 (μπορεί να τερματιστεί νωρίτερα μέσω early stopping)<br>
BATCH = 64 (προσαρμόζεται σε min(BATCH, len(X_trn)) για ασφάλεια)<br>

### 4) Αποτελέσματα μετρικών στο test set

Οι JNJ, PG και XOM εμφανίζουν πολύ χαμηλά MAE/RMSE (JNJ: 1.11/1.60, PG: 1.13/1.58, XOM: 0.95/1.30) και MAPE κάτω από \~1.5%, δείχνοντας ότι το LSTM περιγράφει επαρκώς το μέγεθος των καθημερινών κινήσεων. Αντίθετα, σε τίτλους υψηλού beta η απόδοση υποβαθμίζεται. Η TSLA έχει τα υψηλότερα σφάλματα (MAE 5.31, RMSE 8.18, MAPE 2.85%), ενώ η AMD ακολουθεί (MAE 2.26, MAPE 2.45%). Η SPY εμφανίζει σχετικά υψηλό MAE/RMSE λόγω κλίμακας τιμών, αλλά πολύ χαμηλό MAPE (0.97%), κάτι που υποδηλώνει ότι το σχετικό σφάλμα είναι μικρό παρά την απόλυτη τιμή.

Στις κατευθυντικές μετρικές, οι τιμές είναι συμπαγείς γύρω από το 0.50–0.54, στοιχείο που υποδεικνύει ισορροπημένη αλλά ήπια ικανότητα πρόβλεψης κατεύθυνσης. Η JNJ ξεχωρίζει με το υψηλότερο Hit Rate (0.532) και F1≈0.534, ενώ η SPY εμφανίζει το μεγαλύτερο F1 (0.534) χάρη σε συνδυασμό καλής precision/recall (0.528/0.540). Ο XOM κινείται ομοιόμορφα γύρω από το 0.51, ενώ η AMD υστερεί (F1 0.476). Ένα σταθερό μοτίβο είναι ότι το recall υπερβαίνει οριακά το precision σε όλες τις μετοχές, γεγονός που σημαίνει ότι το LSTM τείνει να πιάνει πραγματικές κινήσεις αλλά έχει ψευδώς θετικές προβλέψεις.

Συνολικά, τα ραβδογράμματα επιβεβαιώνουν ότι το LSTM είναι ιδιαίτερα αποτελεσματικό σε «σταθερούς» ή δομικά κατευθυνόμενους τίτλους (JNJ, PG, XOM) με χαμηλά σφάλματα και άνω του 0.53 F1, ενώ σε high-beta περιβάλλοντα (TSLA, AMD) διατηρεί αξιοπρεπή κατευθυντική επίδοση (\~0.51), αλλά με σαφώς υψηλότερα σφάλματα μεγέθους. * Το **καθεστώς υψηλής μεταβλητότητας** είναι το πιο απαιτητικό περιβάλλον: τα σφάλματα κορυφώνονται (MAE ≈ 3.78, RMSE ≈ 5.74, MAPE ≈ 2.65%) και οι κατευθυντικές μετρικές υποχωρούν (Hit Rate ≈ 0.489, Precision ≈ 0.487, Recall ≈ 0.504, F1 ≈ 0.495). Το LSTM δυσκολεύεται να ανταποκριθεί σε απότομες κινήσεις και αιχμές.

* Στο **καθεστώς χαμηλής μεταβλητότητας** καταγράφονται τα μικρότερα σφάλματα (MAE ≈ 1.12, RMSE ≈ 1.59, MAPE ≈ 0.84%) και ο υψηλότερος Hit Rate (≈ 0.510). Η κατευθυντική επίδοση είναι σταθερή (Precision ≈ 0.519, Recall ≈ 0.533, F1 ≈ 0.526). Το μοντέλο ωφελείται από τις ομαλές, επαναλαμβανόμενες κινήσεις τιμών.

### 6) Διαγράμματα κυλιόμενης μεταβλητότητας - σφάλματος

Παρατηρώντας τα διαγράμματα, το πρότυπο μοτίβο είναι πως όσο αυξάνεται η βραχυπρόθεσμη μεταβλητότητα, διευρύνεται η κατανομή των σφαλμάτων. Ωστόσο, το τελικό μοντέλο (πορτοκαλί) σχηματίζει συστηματικά πυκνότερο σύννεφο κοντά στο 0 και με μικρότερη ουρά σε σχέση με το LSTM (μωβ), δείχνοντας σαφή υπεροχή σε όλο το εύρος μεταβλητότητας.

Στις AAPL και XOM έχουμε ήπια θετική κλίση σφάλματος–μεταβλητότητας. Το LSTM εμφανίζει συχνότερα outliers >5, ενώ το FinalModel παραμένει συμπαγές σε χαμηλά σφάλματα.

Στις TSLA και AMD τα δύο μοντέλα υποφέρουν όταν η μεταβλητότητα ξεπερνά \~0.06–0.08. Παρ’ όλα αυτά, ο όγκος των σφαλμάτων του τελικού μοντέλου μετατοπίζεται χαμηλότερα, μειώνοντας τη μάζα των σφαλμάτων 10–30, αν και παραμένουν να υπάρχουν ακραίες τιμές.
  
Στις SPY και JNJ έχουμε παρόμοια επίδοση με λίγο μεγαλύτερα σφάλματα για το LSTM.
 
Για την PG η διαφορά είναι εντονότερη υπέρ του τελικού μοντέλου, που έχει πολύ συμπαγές νέφος κοντά στο 0.

Συνολικά, το τελικό ΤΤΜ μειώνει καλύτερα τόσο το επίπεδο όσο και τη διασπορά του σφάλματος για όλες τις μετοχές και σε όλες τις ζώνες μεταβλητότητας, με τη μεγαλύτερη ωφέλεια στις μετοχές υψηλού beta (TSLA, AMD) και καθαρή βελτίωση ακόμη και στον δείκτη SPY. Το υπόλοιπο ρίσκου εντοπίζεται στις ακραίες μεταβλητότητες, όπου παραμένουν μεμονωμένα μεγάλα σφάλματα.

### 7) Πίνακας Pearson

Οι τιμές είναι κοντά στο μηδέν ή αρνητικές για τις περισσότερες μετοχές (AAPL −0.27, SPY −0.25, PG −0.20), με μία μόνο οριακά θετική εξαίρεση (XOM 0.06).

Για το LSTM δεν υπάρχει ισχυρή γραμμική σχέση μεταξύ μεταβλητότητας και μεγέθους σφάλματος. 'Οπου η σχέση υπάρχει, είναι ελαφρώς αντιστρόφως ανάλογη. Με απλά λόγια, οι περίοδοι υψηλότερης ημερήσιας μεταβλητότητας δεν συνοδεύονται από συστηματική διόγκωση των σφαλμάτων και, σε ορισμένους τίτλους, τα σφάλματα τείνουν να μειώνονται ελαφρά όταν αυξάνεται η μεταβλητότητα.

Αυτό αντιδιαστέλλεται με τα προηγούμενα μοντέλα. Τόσο το ARIMA όσο και το τελικο ΤΤΜ εμφάνιζαν θετική συσχέτιση (μέτρια για SPY/PG/JNJ, υψηλότερη για TSLA).

Η αδυναμία ανίχνευσης θετικής σχέσης υποδηλώνει ότι το LSTM είναι λιγότερο ευαίσθητο στη μεταβλητότητα καθώς οι μηχανισμοί μνήμης φαίνεται να βοηθούν στην απορρόφηση γρήγορων διακυμάνσεων χωρίς εκτίναξη του σφάλματος.

Αυτό δεν σημαίνει κατ’ ανάγκη ότι το LSTM είναι καλύτερο σε όλες τις ζώνες ρίσκου. Το Pearson συλλαμβάνει μόνο γραμμική τάση και μπορεί να κρύψει παχιές ουρές σφαλμάτων. Στα διαγράμματα διασποράς είδαμε ότι το LSTM εξακολουθεί να εμφανίζει μεμονωμένα μεγάλα σφάλματα, ειδικά σε τίτλους υψηλού beta.

### 8) Οπτικοποίηση τελικής βελτιστοποίησης σε σχέση με το τελικό μοντέλο ΤΤΜ

**1) Θερμικοί χάρτες σφαλμάτων**<br>
Σε όλες τις μετοχές οι διαφορές είναι αρνητικές δηλαδή το LSTM υπερέχει συστηματικά στα σφάλματα. Η μεγαλύτερη απόσταση καταγράφεται στην TSLA (≈−1.64 MAE, −1.95 RMSE, −1.09 MAPE). Σημαντικά κέρδη υπέρ του LSTM εμφανίζονται και σε SPY (≈−0.20, −0.50, −0.08) και AMD (≈−0.41, −0.53, −0.46). Πιο ήπιες αλλά σταθερές βελτιώσεις βλέπουμε σε AAPL, XOM, JNJ και PG (μεγέθους \~−0.1 έως −0.35).

**2) Θερμικοί χάρτες κατευθυνσης**<br>
Θετικό πρόσημο σημαίνει καλύτερη κατεύθυνση για το τελικό μοντέλο. Η εικόνα είναι μικτή.

Στα Recall και F1 το τελικό μοντέλο υπερέχει καθαρά σε AAPL (+0.012 / +0.004), TSLA (+0.010 / +0.005), SPY (+0.025 / +0.009), AMD (+0.021 / +0.015) και PG (+0.010 / +0.002). Αυτό σημαίνει ότι έχει λιγότερες ψευδώς αρνητικές προβλέψεις, ιδιαίτερα σε μετοχές με έντονη δυναμική.

Στα Hit rate & Precision οι διαφορές είναι μικρές και γύρω από το μηδέν, άρα τα δύο μοντέλα έχουν παρόμοια συνολική συχνότητα σωστής κατεύθυνσης και ακρίβεια σημάτων.

Σε XOM και JNJ τα πρόσημα είναι ελαφρώς αρνητικά σε όλες τις κατευθυντικές μετρικές (π.χ. XOM −0.014 hit rate, −0.015 F1), υποδηλώνοντας ένα μικρό πλεονέκτημα του LSTM σε αυτά τα πιο αμυντικά καθεστώτα.

**Συμπέρασμα**<br>
Οι θερμικοί χάρτεσ καταδεικνύουν ότι, ως προς το επίπεδο σφάλματος, το LSTM είναι με μικρές διαφορές ανώτερο του τελικού μοντέλου. Κατευθυντικά, τα δύο μοντέλα είναι κοντά σε hit rate/precision, αλλά το τελικό μοντέλο παρουσιάζει καθαρό πλεονέκτημα σε recall και F1 στις περισσότερες μετοχές, δηλαδή αναγνωρίζει περισσότερες πραγματικές κινήσεις χωρίς μεγάλη θυσία ακρίβειας. Μόνη ουσιαστική εξαίρεση αποτελούν XOM και JNJ, όπου το LSTM διατηρεί μια μικρή υπεροχή στην κατεύθυνση.

### 9) Αποτελέσματα στα παράθυρα γεγονότων

**Κατάρρευση & Ανάκαμψη (AAPL, 2020-03-10).**
Το LSTM συλλαμβάνει πειστικά την ανοδική πορεία πριν την πτώση και την έντονη πτωτική κίνηση του Μαρτίου, αλλά αποκτά χρονική καθυστέρηση όταν αρχίζουν μικρά σκαμπανεβάσματα στην τιμή. Κατά τη διάρκεια της ανάκαμψης ακολουθεί πολύ κοντά, αλλά με μικρή υστέρηση μερικών ημερών στις κορυφές. Το πλάτος είναι ελαφρώς στρογγυλεμένο.

**Αποφόρτιση Μετοχής Υψηλού Beta (TSLA, 2021-01-15).**
Εξαιρετική σύλληψη της γενικής κατεύθυνσης της κίνησης αλλά υπάρχει σαφές πρόβλημα συγχρονισμού με την πραγματική τιμή καθ' όλη τη διάρκεια του παραθύρου.

**Κορύφωση Πετρελαϊκού Κύκλου (XOM, 2022-06-01).**
Πολύ καλή στοίχιση τόσο στη μεσοπρόθεσμη ανοδική κίνηση όσο και στα pullbacks μετά την κορύφωση. Οι αποκλίσεις είναι μικρές και σχεδόν συμμετρικές, ένδειξη ότι το LSTM έχει μάθει το μοτίβο του κύκλου χωρίς σημαντικές χρονικές καθυστερήσεις.

**Πτωτική Αστάθεια με Εναλλαγές Κατεύθυνσης (SPY, 2022-09-15).**
Το μοντέλο αναπαριστά τη διαδοχή πτωτικών ποδιών και βραχύβιων αναπηδήσεων, αλλά εμφανίζει μικρές σφαλματώσεις φάσης. Τα τοπικά υψηλά/χαμηλά έρχονται με καθυστέρηση 1-3 ημερών και το εύρος των μικροκύκλων είναι ελαφρά μειωμένο. Παρ’ όλα αυτά η συνολική καθοδική δομή αποδίδεται πειστικά.

**Περίοδος Χαμηλής Μεταβλητότητας (JNJ, 2019-08-01).**
Εδώ το LSTM κολλάει σχεδόν πάνω στην πραγματική τιμή. Η χαμηλή μεταβλητότητα ευνοεί την ακριβή πρόβλεψη επιπέδων. Οι διαφορές είναι μικροσκοπικές, με σποραδικές καθυστερήσεις μόνο σε απότομες μονοήμερες κινήσεις.

**Μαζική Πώληση Τεχνολογικών Μετοχών (AMD, 2018-10-10).**
Καλή σύλληψη της έντονης ανόδου πριν από τη διόρθωση και της επακόλουθης πτώσης. Όπως και στην TSLA, οι αιχμές εξομαλύνονται. Το κάτω άκρο της πτώσης και τα γρήγορα ριμπάουντ της τιμής αποδίδονται με ήπια υστέρηση και μικρότερη ένταση σε σχέση με την πραγματικότητα.

**Ηρεμία Ανεπηρέαστη από Μακροοικονομικούς Παράγοντες (PG, 2015-06-15).**
Πολύ σταθερή στοίχιση σε όλο το παράθυρο. Το μοντέλο παρακολουθεί τη σταδιακή κάμψη και τη μετέπειτα σταθεροποίηση χωρίς συστηματικά σφάλματα. Αναμένουμε πολύ χαμηλές μετρικές σφάλματος εδώ.

### 11) Αποτελέσματα στρατιγικής - ποσοστιαία διαγράμματα
Ξεκινάμε από 0% κεφάλαιο και σε κάθε βήμα παίρνουμε θέση long όταν το μοντέλο προβλέπει άνοδο και θέση short όταν προβλέπει πτώση. Η καμπύλη κάθε μετοχής δεν είναι η τιμή της, αλλά το αποτέλεσμα του συγκεκριμένου κανόνα πάνω στη μετοχή. Αν η καμπύλη ανεβαίνει, σημαίνει ότι οι προβλέψεις έπιαναν σωστά το πρόσημο των κινήσεων εκείνη την περίοδο, αν κατεβαίνει, οι προβλέψεις ήταν, κατά μέσο όρο, αντίθετες από την πραγματική κίνηση. Οι απότομες βουτιές είναι drawdowns, ενώ ο τελικός ύψος της καμπύλης είναι η συνολική σωρευτική απόδοση.

Η TSLA έπειτα από μια αρχική αστάθεια, μπαίνει σε παρατεταμένη ανοδική φάση με επαναλαμβανόμενες κορυφές >100% και κλείσιμο γύρω στο +110–120%. Η AMD παρουσιάζει παρόμοια συμπεριφορά, με μεγάλα ράλι και εξίσου μεγάλες διορθώσεις, αλλά τερματίζει επίσης ψηλά (≈+110–120%), επιβεβαιώνοντας την προτίμηση του μοντέλου για high-beta περιβάλλοντα. Ο SPY ανεβαίνει σταθερά, χωρίς ακραίες ταλαντώσεις, και καταλήγει κοντά στο +60–70%. Πρόκειται για το πιο καθαρό και ανθεκτικό μονοπάτι για την στρατιγική. Οι αμυντικοί τίτλοι έχουν ηπιότερες επιδόσεις. Η JNJ σκαρφαλώνει μέχρι περίπου +60–70% αλλά υποχωρεί προς το τέλος σε \~+30–40%, ενώ η PG κινείται με μικρές κλίσεις και κλείνει γύρω στο +15–25%. Η XOM είναι καθαρά κυκλική. Μετά από ράλι στο μέσο της περιόδου ακολουθεί παρατεταμένη αποδυνάμωση και μόλις στο τέλος ανακάμπτει κοντά στο μηδέν έως +10%. Τέλος, η AAPL υποφέρει από έντονα drawdowns στο δεύτερο μισό του ορίζοντα και καταλήγει ελαφρά αρνητική, ένδειξη ότι το μοντέλο δεν προσαρμόστηκε έγκαιρα στη μεταστροφή της μετοχής.

### 12) Αποτελέσματα στρατιγικής - διαγράμματα Sharp, Total Return, Max Drawdown

Τα τρία ραβδογράμματα, για κάθε μετοχή χωριστά, συνοψίζουν την απόδοση της ίδιας ακριβώς δοκιμαστικής στρατηγικής που τρέχουμε πάνω στις προβλέψεις του μοντέλου: 1) το Sharpe ratio μετρά πόση απόδοση ανά μονάδα κινδύνου πέτυχε η στρατηγική, 2) το Total return είναι το συνολικό κερδισμένο ποσοστό στο τέλος της περιόδου, ενώ 3) το Max drawdown είναι το μεγαλύτερο ποσοστό πτώσης της καμπύλης πλούτου της στρατηγικής, άρα δείχνει τον χειρότερο ενδιάμεσο «κραδασμό» που θα χρειαζόταν να αντέξουμε.

Στο Sharpe Ratio το LSTM αναδεικνύει καθαρό νικητή τον SPY (≈0.04), άρα η στρατηγική παράγει την υψηλότερη απόδοση ανά μονάδα κινδύνου όταν εφαρμόζεται στον ευρύ δείκτη αγοράς. Οι TSLA και AMD ακολουθούν (≈0.02), υποδεικνύοντας ελκυστική, αλλά όχι κορυφαία σχέση ρίσκου-απόδοσης. Οι PG (≈0.01) και XOM/ JNJ (\~0.00) κινούνται οριακά πάνω από το μηδέν, με AAPL να εμφανίζει ελαφρά αρνητική τιμή. Η εικόνα επιβεβαιώνεται και στα Total Returns. Οι TSLA και AMD γράφουν τριψήφια κέρδη (≈112.7% και ≈110.8% αντίστοιχα), ο SPY ακολουθεί με ≈71.0%, ενώ οι PG** και XOM αποδίδουν μέτρια (≈21.7% και ≈10.7%). Η JNJ κινείται οριακά θετικά (≈6.0%) και η AAPL καταγράφει μικρή αρνητική επίδοση (≈−3.2%). Στο Max Drawdown για TSLA και AMD οι μέγιστες υποχωρήσεις είναι εξαιρετικά μεγάλες (≈141% και ≈130%), ένδειξη έντονης ευπάθειας σε απότομες αντιστροφές και συσσώρευσης ζημιών όταν το σήμα κολλάει σε λάθος πλευρά της αγοράς. Ο SPY παρουσιάζει τον μικρότερο κραδασμό (≈39%) μεταξύ των υψηλών αποδόσεων, ενώ οι PG και JNJ έχουν επίσης σχετικά περιορισμένα drawdowns (≈34% και ≈69%), σε αντιστοιχία με τον πιο αμυντικό τους χαρακτήρα. Η AAPL εμφανίζει μεγάλο drawdown (≈106%), στοιχείο συμβατό με τη μειονεκτική συνολική απόδοση.
