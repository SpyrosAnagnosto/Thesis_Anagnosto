#### 2) Συσταδοποίηση βάση σφαλμάτων
Στο προηγούμενο βήμα μελετήσαμε την μεταφερσιμότητα των μοντέλων κοιτώντας τις τιμές των αποτελεσμάτων αυτές καθ' αυτές. Σε αυτό το βήμα παρατηρούμε τα αποτελέσματα πιο συγκεντρωτικά με σκοπό να καταλήξουμε σε μια πιο πλήρη εικόνα. 
Το πρώτο βήμα είναι, από τους θερμικούς χάρτες να κατασκευάσουμε χάρτες συστάδων (clustermaps). Ο χάρτης συστάδων παίρνει τον θερμικό χάρτη και ομαδοποιεί παρόμοιες γραμμές και στήλες. 

Όταν γραμμές ομαδοποιούνται, σημαίνει ότι τα μοντέλα που έχουν εκπαιδευτεί σε αυτές τις μετοχές κάνουν παρόμοια σφάλματα σε άλλες μετοχές. Αυτό δείχνει ότι τα μοντέλα μαθαίνουν από τα δεδομένα με τρόπο που γενικεύεται παρόμοια. Κάποιες φορές, η συσταδοποίηση ομαδοποιεί μετοχές που δεν φαίνεται να σχετίζονται ούτε από άποψη κλάδου, ούτε μεταβλητότητας. Αυτό μπορεί να υποδηλώνει υπόγειες ομοιότητες στη δυναμική της χρονοσειράς. Μοντέλα που βρίσκονται σε απομονωμένες συστάδες  τείνουν να συμπεριφέρονται πολύ διαφορετικά. Όταν παρατηρείται η ίδια συσταδοποίηση σε RMSE, MAE και MAPE, ενισχύεται η εμπιστοσύνη ότι οι σχέσεις είναι πραγματικές και όχι στατιστικός θόρυβος. Η συνέπεια αυτή σημαίνει ότι τα μοντέλα είναι παρόμοια όχι μόνο ως προς το συνολικό σφάλμα, αλλά και ως προς τη δομή και φύση των σφαλμάτων.

Αν δύο στήλες ανήκουν στην ίδια συστάδα, σημαίνει ότι τα περισσότερα μοντέλα κάνουν παρόμοια σφάλματα όταν προβλέπουν αυτές τις δύο μετοχές.
Έτσι, συσταδοποίηση στις στήλες δείχνει αν υπάρχει κάποιο συστηματικό μοτίβο σφάλματος για συγκεκριμένους στόχους.Ορισμένοι στόχοι μπορεί να δέχονται χαμηλά σφάλματα από όλα τα μοντέλα, και έτσι να συγκεντρώνονται σε συστάδες εύκολων στόχων. Άλλοι μπορεί να συσχετίζονται ως δύσκολοι στόχοι, ιδιαίτερα αν είναι ευμετάβλητοι. Αυτή η ομαδοποίηση βοηθά στο να αποφασίσουμε ποιες μετοχές χρειάζονται πιο σύνθετα ή προσαρμοσμένα μοντέλα. Επίσης, αν μια ολόκληρη συστάδα στόχων είναι συνεχώς δύσκολη στην πρόβλεψη, ίσως χρειάζεται να επανεξετάσουμε τις εισόδους του μοντέλου.

Από την πλευρά των στηλών στο διάγραμμα του RMSE φαίνεται πως η TSLA είναι απομονωμένη. Η AMD είναι ομαδοποιημένη με την SPY και όχι με τις AAPL και TSLA. Αφού το RMSE δίνει έμφαση σε μεγάλα σφάλματα, SPY και AMD έχουν παρόμοιο αριθμό μεγάλων σφαλμάτων.  Όμως η SPY στο διάγραμμα του MAE είναι απομονωμένη δείχνοτας πως στην πρόβλεψή της είναι δύσκολος στόχος από την σκοπιά των τυπικών αποκλίσεων. Ως προς το MAPE η XOM συσταδοποιείται με τις TSLA και AMD, δείχνοντας ότι ποσοστιαία σε αυτές τις μετοχές υπάρχει δυσκολία στην πρόβλεψη. Από τα παραπάνω καταλήγουμε στο συμπέρασμα πως TSLA, SPY, AMD και ποσοστιαία XOM χρειάζονται πιο προσαρμοσμένα μοντέλα.

**Γραμμές**
- Γρουπ 1: TSLA, JNJ
- Γρουπ 2: AMD, AAPL
- Γρουπ 3: PG, XOM, SPY  

Και στα τρία διαγράμματα η συσταδοποίηση των γραμμών είναι η ίδια.  Μιά ενδιαφέρουσα παρατήρηση είναι ότι η JNJ δεν ανήκει στην ομάδα των SPY, XOM και PG. Αντιθέτως, η JNJ συσχετίζεται πρώτα με την TSLA, και στη συνέχεια αυτή η υπο-ομάδα συγχωνεύεται με τις AAPL και AMD. Αυτή η συσταδοποίηση δείχνει ότι τα μοντέλα που εκπαιδεύονται στην JNJ έχουν προφίλ γενίκευσης πιο κοντά σε ασταθείς μετοχές. Θα μπορόυσαμε να χρησιμοποιήσουμε την JNJ για την πρόβλεψη της TSLA χωρίς να θπάρχει κίνδυνος υπερπροσαρμογής. Οι AAPL και AMD σχηματίζουν την επόμενη κοντινή ομάδα με παρόμοιες επιδόσεις γενίκευσης, οπότε μοντέλα που εκπαιδεύονται σε AAPL μπορούν θα βρουν μοτίβα και για την πρόβλεψη της AMD . Αντίθετα, οι SPY, PG και XOM συγκεντρώνονται μαζί, γεγονός που επιβεβαιώνει πως τα μοντέλα που εκπαιδεύονται σε αυτές έχουν υψηλή μεταφερσιμότητα και γενική αξιοπιστία όταν εφαρμόζονται σε άλλες μετοχές.

**Συμπέρασμα:**
Από την μελέτη αυτής της συσταδοποίησης καταλήγουμε στο συμπέρασμα πως θα μπορούσαμε να φτιάξουμε ένα κοινό μοντέλο εκπαίδευσης με SPY, JNJ, AAPL.

#### 3) Γράφοι μεταφερσιμότητας κατεύθυνσης
Για να μελετήσουμε την ποιότητα της μεταφερσιμότητας στην κατεύθυνση χρησιμοποιούμε γράφους μεταφερσιμότητας. Σε αυτούς του γράφους κάθε κόμβος αντιπροσοπεύει μια μετοχή που μπορεί να χρησιμοποιηθεί είτε ως σύνολο για εκπαίδευση, είτε ως στόχος για πρόβλεψη. Οπότε για κάθε δυάδα κόμβο υπάρχουν δύο ακμές, μία εισόδου και μία εξόδου. Αυτές τις ακμές τις ενώνουμε παίρνοντας έναν μέσο όρο. Στη συνέχεια τις χρωματίζουμε με βάση την τιμή αυτού του μέσου όρου. Πιο παχιές ή πιο σκούρες ακμές να σημαίνουν χειρότερη απόδοση. Σκοπός μας είναι πάλι να βρούμε αν μετοχές εκτός της SPY θα μπορέσουν να μας βοηθήσουν σε καλύτερες προβλέψεις.

<img src="results/graphs.png" alt="Chart"/>


Στο γράφημα του **Hit Rate**, η XOM παρουσιάζει ισχυρή ικανότητα μεταφοράς, καθώς έχει τις περισσότερες ανοιχτόχρωμες συνδέσεις. Για παράδειγμα, πετυχαίνει hit rate 0.496 στην TSLA, 0.492 στην SPY και 0.504 στην PG. Η SPY επίσης έχει ικανοποιητική μεταφορά, με hit rates όπως SPY → TSLA στο 0.492 και SPY → PG στο 0.490.

Στο **Recall** κυριαρχεί η SPY με SPY → AMD στο 0.562, AAPL → JNJ στο 0.557, SPY → AAPL στο 0.552. Επίσης έχουμε είναι η XOM → PG με τιμή 0.575 και AAPL → AMD στο 0.551. Αυτά δείχνουν ότι οι SPY, AAPL και XOM είναι ισχυρές πηγές μοντέλων που διατηρούν καλή απόδοση σε false negatives. 

Μεταξύ των ισχυρότερων ακμών στο **Precision** συγκαταλέγονται οι SPY → PG με τιμή precision 0.530, η υψηλότερη στο γράφημα, καθώς και οι SPY → TSLA (0.525), AMD → SPY (0.520), XOM → SPY (0.523) και XOM → JNJ (0.519). 

Στο **F1** η SPY έχει απόλυτη υπεροχή. Συγκεκριμένα, η μεταφορά από SPY προς PG επιτυγχάνει F1-score 0.550. Οι SPY → AMD (0.549) και SPY → TSLA (0.541).

**Συμπέρασμα:**
Από τα αποτελέσματα αυτά καταλήγουμε στο συμπέρασμα πως στο μοντέλο εκπαίδευσης με συνδυασμό πολλών μετοχών που καταλήξαμε παραπανω (SPY, JNJ, AAPL), μπορούμε να εντάξουμε και την XOM καθώς έχει καλή μεταφερσιμότητα όσο αφορά την κατεύθυνση.
